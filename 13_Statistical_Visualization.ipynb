# Topic 1 – Histogram

## Step 1 – Understand the Concept

A histogram is a statistical visualization used to understand the
distribution of numerical data.

It divides numerical values into intervals called bins and shows
how many observations fall within each interval.

A histogram helps us understand:

- Where the data is concentrated
- How widely the data is spread
- Whether the distribution is symmetric or skewed
- Whether there may be unusual values

### Purpose

The main purpose of a histogram is to understand the distribution
of a numerical variable.

### When to Use

A histogram is useful when:

- The data is numerical.
- We want to understand its distribution.
- We want to identify skewness.
- We want to observe the spread of values.
- We want to identify possible unusual observations.

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose we collect the ages of customers visiting a shopping website.

If most customers are between 20 and 30 years old, the histogram
will have taller bars around that age range.

This helps us understand the age distribution without looking at
every individual value.

### Numerical Example

Consider the following ages:

18, 20, 21, 22, 24, 25, 26, 27, 30, 32

We can divide them into intervals:

18–21
22–25
26–29
30–33

The histogram counts how many observations fall into each interval.

### Business Example

An e-commerce company can use a histogram to understand the
distribution of customer order values.

For example, the company may discover that most orders are between
₹500 and ₹1,500.

This information can help the business understand customer
purchasing behavior.

### AI/ML Example

In Machine Learning, histograms are commonly used during
Exploratory Data Analysis.

For example, before training a house-price prediction model,
we can visualize the distribution of house prices.

This can help identify skewness, unusual values, and the general
shape of the target variable.


## Step 3 – Implement the Concept

```python
import matplotlib.pyplot as plt

ages = [
    18, 20, 21, 22, 24,
    25, 26, 27, 28, 30,
    31, 32, 35, 36, 40,
    42, 45, 50, 52, 55
]

plt.figure(figsize=(8, 5))

plt.hist(
    ages,
    bins=6,
    edgecolor="black"
)

plt.title("Age Distribution")
plt.xlabel("Age")
plt.ylabel("Frequency")

plt.show()
```

## Step 3 – Implement the Concept

### What Does the Program Do?

The program creates a histogram using a list of customer ages.

The `plt.hist()` function groups the ages into six intervals called
bins and counts how many observations are present in each interval.

### Statistical Concept Demonstrated

The program demonstrates the distribution and frequency of numerical
data.

The height of each bar represents the number of observations in that
age interval.

### How to Interpret the Output

A taller bar means that more people belong to that particular age
range.

A shorter bar means fewer observations are present in that range.

The overall shape of the histogram helps us understand whether the
data is concentrated, spread out, or skewed.

### AI/ML Usage

Histograms are useful during Exploratory Data Analysis before
Machine Learning.

They can help us:

- Understand feature distributions
- Identify skewness
- Identify possible outliers
- Decide whether preprocessing or transformation may be useful

## Advantages

- Easy to understand.
- Shows the distribution clearly.
- Helps identify skewness and spread.
- Useful during Exploratory Data Analysis.

## Limitations

- Only suitable for numerical data.
- The appearance depends on the number of bins.
- Individual data values are not directly visible.

# Topic 2 – Bar Chart

## Step 1 – Understand the Concept

A bar chart is used to compare values across different categories.

Each category is represented by a bar, and the height or length of
the bar represents its value.

Unlike a histogram, the categories in a bar chart are separate
groups.

### Purpose

The purpose of a bar chart is to compare different categories.

### When to Use

Use a bar chart when:

- Comparing categories.
- Working with categorical data.
- Comparing counts or summarized values.
- Comparing sales, revenue, or quantities.

## Step 2 – Demonstrate the Concept

### Real-World Example

A shop wants to compare the number of products sold in different
product categories.

### Numerical Example

Suppose:

Laptop = 120
Mobile = 200
Tablet = 80
Watch = 150

A bar chart represents each product as a separate bar.

### Business Example

An e-commerce company can compare sales of different products.

This helps management identify which products have higher or lower
sales.

### AI/ML Example

A bar chart can be used to visualize class frequencies in a
classification dataset.

For example:

- Cat
- Dog
- Bird

This can help identify whether the dataset is balanced or imbalanced.

## Step 3 – Implement the Concept

```python
import matplotlib.pyplot as plt

products = ["Laptop", "Mobile", "Tablet", "Watch"]
sales = [120, 200, 80, 150]

plt.figure(figsize=(8, 5))

plt.bar(products, sales)

plt.title("Product Sales")
plt.xlabel("Product")
plt.ylabel("Number of Sales")

plt.show()
```



### What Does the Program Do?

The program creates a bar chart showing the number of sales for
four products.

### Statistical Concept Demonstrated

The visualization demonstrates comparison between categorical groups.

### How to Interpret the Output

Each bar represents one product.

A taller bar means higher sales.

In this example, Mobile has the highest number of sales.

### AI/ML Usage

Bar charts can be used to inspect class distributions and categorical
feature frequencies before Machine Learning.

## Advantages

- Simple and easy to understand.
- Excellent for category comparison.
- Works well with categorical data.

## Limitations

- Too many categories can make the chart difficult to read.
- Does not show numerical distributions.
- Not suitable for continuous distributions.

# Topic 3 – Line Chart

## Step 1 – Understand the Concept

A line chart connects data points using lines to show changes
across an ordered sequence.

It is especially useful for time-series data.

### Purpose

The main purpose is to identify trends, increases, decreases,
and changes over time.

### When to Use

Use a line chart when:

- Data has a natural order.
- Data represents time.
- You want to identify trends.
- You want to compare changes over time.

## Step 2 – Demonstrate the Concept

### Real-World Example

We can track daily temperature for one week.

The line chart shows whether temperature is increasing or decreasing.

### Numerical Example

Monthly revenue:

January = 20
February = 25
March = 23
April = 30
May = 35
June = 40

Connecting these values shows the revenue trend.

### Business Example

A company can visualize monthly revenue to identify sales growth
or decline.

### AI/ML Example

Line charts can visualize:

- Training loss
- Validation loss
- Accuracy
- Model performance over epochs


## Step 3 – Implement the Concept


```python
import matplotlib.pyplot as plt

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]
revenue = [20, 25, 23, 30, 35, 40]

plt.figure(figsize=(8, 5))

plt.plot(months, revenue, marker="o")

plt.title("Monthly Revenue")
plt.xlabel("Month")
plt.ylabel("Revenue")

plt.show()
```



### What Does the Program Do?

The program plots monthly revenue and connects each month's value
with a line.

### Statistical Concept Demonstrated

It demonstrates how a numerical variable changes over an ordered
sequence.

### How to Interpret the Output

The upward movement indicates increasing revenue.

There is a small decrease between February and March.

### AI/ML Usage

Line charts can be used to monitor Machine Learning model training,
such as loss and accuracy across epochs.

## Advantages

- Clearly shows trends.
- Useful for time-series data.
- Shows increases and decreases easily.

## Limitations

- Too many lines can make the chart confusing.
- Not suitable for unrelated categories.
- Connecting values may imply continuity where it does not exist.

# Topic 4 – Scatter Plot

## Step 1 – Understand the Concept

A scatter plot displays individual observations as points on two
numerical axes.

It helps us understand the relationship between two numerical
variables.

It can reveal:

- Positive relationships
- Negative relationships
- Clusters
- Outliers
- Possible patterns

### Purpose

The purpose is to investigate the relationship between two numerical
variables.

### When to Use

Use a scatter plot when:

- Both variables are numerical.
- You want to identify a relationship.
- You want to identify clusters or outliers.

## Step 2 – Demonstrate the Concept

### Real-World Example

We can compare the number of hours students study with their exam
marks.

### Numerical Example

Study hours:

1, 2, 3, 4, 5, 6

Marks:

45, 50, 58, 65, 75, 85

As study hours increase, marks generally increase.

### Business Example

A company can compare advertising expenditure with sales.

This can help analysts investigate whether higher advertising
spending is associated with higher sales.

### AI/ML Example

Scatter plots are useful during Exploratory Data Analysis to
understand relationships between features and the target variable.


## Step 3 – Implement the Concept

```python
import matplotlib.pyplot as plt

study_hours = [1, 2, 3, 4, 5, 6, 7, 8]
marks = [45, 50, 55, 62, 68, 75, 82, 90]

plt.figure(figsize=(8, 5))

plt.scatter(study_hours, marks)

plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")

plt.show()
```



### What Does the Program Do?

The program creates points using study hours and marks.

Each point represents one observation.

### Statistical Concept Demonstrated

It demonstrates the relationship between two numerical variables.

### How to Interpret the Output

The points generally move upward.

This indicates a positive relationship between study hours and marks
in this example.

However, a scatter plot does not prove that one variable causes
the other.

### AI/ML Usage

Scatter plots help identify feature relationships, clusters,
outliers, and possible linear or nonlinear patterns.

## Advantages

- Shows relationships clearly.
- Helps identify trends.
- Can reveal clusters and outliers.

## Limitations

- Can become crowded with large datasets.
- Does not prove causation.
- A basic scatter plot directly shows only two variables.

# Topic 5 – Pie Chart

## Step 1 – Understand the Concept

A pie chart represents parts of a whole using sections of a circle.

Each section represents the proportion of one category.

### Purpose

The purpose is to show how a total is divided among categories.

### When to Use

Use a pie chart when:

- Categories represent parts of a common total.
- There are only a few categories.
- Proportions are more important than exact comparisons.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company can show the percentage of customers using different
payment methods.

### Numerical Example

Suppose payment methods are:

UPI = 45%
Card = 30%
Cash = 15%
Net Banking = 10%

Together they represent 100% of payments.

### Business Example

A business can visualize the percentage contribution of different
product categories to total sales.

### AI/ML Example

A pie chart can be used for a small classification dataset to
show the proportion of each class.

For example:

- Cat
- Dog
- Bird


## Step 3 – Implement the Concept

```python
import matplotlib.pyplot as plt

methods = ["UPI", "Card", "Cash", "Net Banking"]
values = [45, 30, 15, 10]

plt.figure(figsize=(7, 7))

plt.pie(
    values,
    labels=methods,
    autopct="%1.1f%%"
)

plt.title("Payment Method Distribution")

plt.show()
```



### What Does the Program Do?

The program creates a pie chart using payment method percentages.

### Statistical Concept Demonstrated

It demonstrates part-to-whole relationships.

### How to Interpret the Output

Each slice represents one payment method.

The largest slice represents UPI, which has the highest proportion
in this example.

### AI/ML Usage

Pie charts can be useful for quickly checking class proportions
when there are only a few classes.

## Advantages

- Easy to understand for simple proportions.
- Shows percentage contribution.
- Useful for small numbers of categories.

## Limitations

- Similar-sized slices are difficult to compare.
- Too many categories make it confusing.
- Exact comparisons are usually easier with a bar chart.

# Topic 6 – Density Plot

## Step 1 – Understand the Concept

A density plot is used to visualize the estimated distribution of
continuous numerical data.

Instead of showing separate histogram bars, it creates a smooth
curve representing estimated data density.

A common method used to create a density plot is
Kernel Density Estimation (KDE).

### Purpose

The purpose is to understand the shape and concentration of a
numerical distribution.

### When to Use

Use a density plot when:

- Data is continuous numerical data.
- You want a smooth distribution.
- You want to compare distributions between groups.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company can visualize the distribution of customer spending.

The peaks of the curve show where observations are more concentrated.

### Numerical Example

Customer spending:

20, 25, 30, 35, 40, 45, 50, 55, 60

A density plot creates a smooth curve representing where spending
values are concentrated.

### Business Example

A business can compare spending distributions between:

- New customers
- Returning customers

### AI/ML Example

Density plots can help understand feature distributions before
Machine Learning and can reveal skewness or multiple peaks.


## Step 3 – Implement the Concept

```python
import pandas as pd
import matplotlib.pyplot as plt

spending = [
    20, 25, 30, 28, 35,
    40, 42, 45, 50, 52,
    55, 58, 60, 62, 65,
    70, 72, 75, 80, 85
]

series = pd.Series(spending)

series.plot(kind="density")

plt.title("Customer Spending Density")
plt.xlabel("Spending")

plt.show()
```



### What Does the Program Do?

The program converts the spending values into a Pandas Series
and creates a density plot.

### Statistical Concept Demonstrated

It demonstrates an estimated smooth distribution of numerical data.

### How to Interpret the Output

Higher parts of the curve represent areas where observations are
more concentrated.

The shape helps us understand:

- Center
- Spread
- Skewness
- Possible multiple peaks

### AI/ML Usage

Density plots can help identify feature distributions and determine
whether transformations may be useful.

## Advantages

- Provides a smooth distribution.
- Useful for comparing groups.
- Shows distribution shape clearly.

## Limitations

- Depends on KDE smoothing/bandwidth.
- Does not show exact individual observations.
- Can be difficult to interpret with very small datasets.

# Topic 7 – Box Plot

## Step 1 – Understand the Concept

A box plot is a statistical visualization that summarizes numerical
data using quartiles and the median.

It usually shows:

- First Quartile (Q1)
- Median
- Third Quartile (Q3)
- Whiskers
- Potential outliers

The Interquartile Range is:

IQR = Q3 - Q1

A common rule identifies values below:

Q1 - 1.5 × IQR

or above:

Q3 + 1.5 × IQR

as potential outliers.

### Purpose

The purpose is to understand the center, spread, and potential
outliers of numerical data.

### When to Use

Use a box plot when:

- Comparing distributions between groups.
- Identifying potential outliers.
- Understanding median and spread.

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose a company records employee salaries.

A box plot can show the median salary, spread of salaries,
and potential unusually high or low salaries.

### Numerical Example

Salaries:

20, 22, 23, 24, 25, 26, 27, 28, 30, 32, 50

The value 50 may appear as a potential high outlier.

### Business Example

A company can compare salaries across different departments
using multiple box plots.

### AI/ML Example

Box plots are useful during EDA for detecting potential outliers
in numerical features before Machine Learning.


## Step 3 – Implement the Concept

```python
import matplotlib.pyplot as plt

salaries = [
    20, 22, 23, 24, 25,
    26, 27, 28, 30, 32, 50
]

plt.figure(figsize=(7, 5))

plt.boxplot(salaries)

plt.title("Salary Distribution")
plt.ylabel("Salary")

plt.show()
```



### What Does the Program Do?

The program creates a box plot using salary data.

### Statistical Concept Demonstrated

It demonstrates quartiles, median, spread, and potential outliers.

### How to Interpret the Output

The line inside the box represents the median.

The box represents the middle 50% of the observations.

The whiskers represent the range according to the box plot rule.

Points beyond the whiskers may be potential outliers.

### AI/ML Usage

Box plots can help identify unusual values in numerical features
during data preprocessing.

An outlier should not automatically be deleted. It should be
investigated based on the context.

## Advantages

- Shows median and spread quickly.
- Helps identify potential outliers.
- Useful for comparing groups.

## Limitations

- Does not show the complete distribution shape.
- Individual observations are mostly hidden.
- Potential outliers are not automatically data errors.

# Topic 8 – Violin Plot

## Step 1 – Understand the Concept

A violin plot combines information from a box plot with a
density-based representation of the data distribution.

It can show:

- Median
- Spread
- Distribution shape
- Density
- Potential unusual observations

### Purpose

The purpose is to compare the distribution of numerical data
between groups.

### When to Use

Use a violin plot when:

- Comparing multiple groups.
- Understanding distribution shape.
- A box plot alone does not provide enough distribution detail.

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose a company wants to compare employee salaries across
three departments.

A violin plot can show both the central value and the shape
of salary distributions.

### Numerical Example

Group A:

20, 22, 23, 25, 26, 28, 30

Group B:

30, 32, 35, 36, 38, 40, 42

The two distributions have different centers and ranges.

### Business Example

An e-commerce company can compare customer spending distributions
across customer segments.

### AI/ML Example

A violin plot can compare a numerical feature across different
target classes.


## Step 3 – Implement the Concept

```python
import matplotlib.pyplot as plt

group_a = [20, 22, 23, 25, 26, 28, 30]
group_b = [30, 32, 35, 36, 38, 40, 42]
group_c = [25, 27, 28, 29, 30, 31, 32]

plt.figure(figsize=(8, 5))

plt.violinplot(
    [group_a, group_b, group_c],
    showmedians=True
)

plt.xticks(
    [1, 2, 3],
    ["Group A", "Group B", "Group C"]
)

plt.title("Group Distribution Comparison")
plt.ylabel("Value")

plt.show()
```



### What Does the Program Do?

The program creates three violin plots representing three groups.

The median is displayed using `showmedians=True`.

### Statistical Concept Demonstrated

The visualization demonstrates both the distribution shape and
central tendency of each group.

### How to Interpret the Output

The width of the violin indicates estimated data density.

Wider sections indicate that more observations are concentrated
around those values.

### AI/ML Usage

Violin plots can be used during EDA to compare feature distributions
between different target classes.

## Advantages

- Shows distribution shape.
- Useful for group comparison.
- Provides more distribution information than a basic box plot.

## Limitations

- Can be more difficult to interpret.
- Density estimation depends on smoothing.
- Can be misleading with very small datasets.

# Topic 9 – Pair Plot

## Step 1 – Understand the Concept

A pair plot displays relationships between multiple numerical
variables at the same time.

It generally contains:

- Scatter plots between pairs of variables.
- Distribution plots on the diagonal.

It provides a quick overview of relationships in a dataset.

### Purpose

The purpose is to explore relationships between multiple variables
during Exploratory Data Analysis.

### When to Use

Use a pair plot when:

- There are several numerical features.
- You want to inspect pairwise relationships.
- You want a quick EDA overview.

## Step 2 – Demonstrate the Concept

### Real-World Example

For students, we can analyze:

- Study Hours
- Attendance
- Sleep Hours
- Marks

A pair plot can display relationships between all these variables.

### Numerical Example

For example, we may observe that:

Study Hours increase → Marks generally increase.

Attendance increases → Marks generally increase.

### Business Example

An e-commerce company can analyze:

- Customer age
- Number of purchases
- Spending
- Website visits

The pair plot helps analysts inspect relationships between these
variables.

### AI/ML Example

Pair plots can be used during EDA to understand relationships
between features and identify possible clusters or patterns before
training a Machine Learning model.


## Step 3 – Implement the Concept

```python
%pip install seaborn

import seaborn as sns
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5, 6],
    "Attendance": [60, 65, 70, 75, 80, 85],
    "Marks": [45, 50, 58, 65, 75, 85],
    "Sleep_Hours": [8, 7, 7, 6, 6, 5]
})

sns.pairplot(data)

plt.show()
```



### What Does the Program Do?

The program creates a pair plot using four numerical features.

It displays scatter plots between feature pairs and distributions
on the diagonal.

### Statistical Concept Demonstrated

It demonstrates pairwise relationships between multiple numerical
variables.

### How to Interpret the Output

Each off-diagonal graph shows the relationship between two variables.

The diagonal shows the distribution of each variable.

For example, we can inspect the relationship between Study_Hours
and Marks.

### AI/ML Usage

Pair plots are useful during EDA to identify:

- Feature relationships
- Possible clusters
- Outliers
- Feature-target patterns

## Advantages

- Provides a broad overview.
- Useful for Exploratory Data Analysis.
- Shows many pairwise relationships.

## Limitations

- Becomes difficult to read with many features.
- Can be computationally expensive for large datasets.
- Mainly shows pairwise relationships.

# Topic 10 – Heatmap

## Step 1 – Understand the Concept

A heatmap represents numerical values using visual intensity.

It is commonly used to visualize a matrix of values.

One important application is displaying a correlation matrix.

### Purpose

The purpose is to identify patterns and compare many numerical
values quickly.

### When to Use

Use a heatmap when:

- Working with a matrix.
- Visualizing correlations.
- Comparing many numerical values.
- Looking for strong or weak relationships.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company can create a heatmap showing sales across different
products and months.

Higher values can be visually distinguished from lower values.

### Numerical Example

Suppose we have correlation values:

Study Hours ↔ Marks = 0.95

Attendance ↔ Marks = 0.85

Sleep Hours ↔ Marks = -0.70

A heatmap makes these relationships easier to identify visually.

### Business Example

A business can create a heatmap of sales by:

- Product
- Month

This can help identify periods and products with higher sales.

### AI/ML Example

A correlation heatmap can be used during EDA to identify highly
related features and investigate possible multicollinearity.


## Step 3 – Implement the Concept

```python
%pip install seaborn

import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

data = pd.DataFrame({
    "Study_Hours": [1, 2, 3, 4, 5, 6],
    "Attendance": [60, 65, 70, 75, 80, 85],
    "Marks": [45, 50, 58, 65, 75, 85],
    "Sleep_Hours": [8, 7, 7, 6, 6, 5]
})

correlation = data.corr()

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation,
    annot=True,
    fmt=".2f"
)

plt.title("Correlation Heatmap")

plt.show()
```



### What Does the Program Do?

First, the program creates a DataFrame containing numerical features.

Then `data.corr()` calculates the correlation between each pair
of variables.

Finally, `sns.heatmap()` displays the correlation matrix visually.

### Statistical Concept Demonstrated

The program demonstrates correlation and matrix-based visualization.

### How to Interpret the Output

Values closer to +1 indicate a strong positive linear relationship.

Values closer to -1 indicate a strong negative linear relationship.

Values closer to 0 indicate a weak or no linear relationship.

### AI/ML Usage

Correlation heatmaps are commonly used during EDA to identify:

- Strong relationships
- Potential redundant features
- Possible multicollinearity
- Feature relationships with a numerical target

Correlation does not prove causation.

## Advantages

- Makes matrix patterns easy to identify.
- Useful for correlation analysis.
- Can display many relationships compactly.

## Limitations

- Too many variables can make the heatmap difficult to read.
- Correlation only describes the measured relationship, not causation.
- Interpretation depends on the chosen correlation measure.
