# Topic 1 - Mean


## Step 1 – Understand the Concept

Mean is the average value of a set of numerical data.

It is calculated by adding all the values and dividing the total by the number of values.

In simple words, mean tells us the central or average value of the given data.

### Formula

Mean = Sum of all values / Number of values

Mean = ΣX / n

Where:

- X = Each value
  
- ΣX = Sum of all values
 
- n = Number of values

## Step 2 – Demonstrate the Concept

### Numerical Example

Suppose the daily sales of a shop are:

100, 150, 200, 250, 300

Mean = (100 + 150 + 200 + 250 + 300) / 5

Mean = 1000 / 5

Mean = 200

Therefore, the average daily sales are 200 units.

### Business Example

A shop owner wants to understand the average number of products sold per day.

If the sales for five days are 100, 150, 200, 250, and 300 units, the mean is 200 units.

This helps the business understand its typical daily sales level.

### AI/ML Example

Suppose an ML dataset contains customer ages:

20, 25, 30, 35, 40

The mean age is 30.

Mean can be used during Exploratory Data Analysis (EDA) to understand the typical value of a numerical feature.

### How does this demonstrate the concept?

The raw sales data is converted into one representative value, 200.
This representative value helps us understand the center of the data.

## Step 3: Implement the Concept

```python
import numpy as np

sales = np.array([100, 150, 200, 250, 300])

mean_sales = np.mean(sales)

print("Mean Sales:", mean_sales)
```

### Interpretation

### Output

Mean Sales: 200.0

### What does the program do?

The program stores the sales values in a NumPy array and calculates their average using np.mean().

### What statistical concept does it demonstrate?

It demonstrates the calculation of the arithmetic mean.

### How do we interpret the output?

The average daily sales are 200 units.

### Where is it used in AI/ML?

Mean is commonly used during data analysis and can also be used for handling missing numerical values through mean imputation.

# Topic 2 – Median

## Step 1 – Understand the Concept

Median is the middle value of a dataset after arranging the values in ascending or descending order.

If the number of values is odd, the middle value is the median.

If the number of values is even, the median is the average of the two middle values.

Median is useful when the data contains extreme values or outliers because it is less affected by them than the mean.

### Formula

For an odd number of values:

Median = Middle value

For an even number of values:

Median = (Two middle values) / 2



## Step 2 – Demonstrate the Concept

### Numerical Example

Data:

10, 20, 30, 40, 50

The middle value is 30.

Therefore:

Median = 30

For an even number of values:

10, 20, 30, 40

The two middle values are 20 and 30.

Median = (20 + 30) / 2

Median = 25

### Business Example

Suppose employee salaries are:

25,000, 30,000, 35,000, 40,000, 5,00,000

The very high salary of 5,00,000 is an extreme value.

The mean will be strongly affected by this value, while the median remains 35,000.

Therefore, median can give a better representation of the typical salary in this example.

### AI/ML Example

Suppose a customer income feature contains:

20,000, 25,000, 30,000, 35,000, 10,00,000

The extreme income value can affect the mean.

Median can be used to understand the central value with less influence from the outlier.

### How does this demonstrate the concept?

The median identifies the middle position of the ordered data and is less sensitive to extreme values.


## Step 3: Implement the Concept

```python
import numpy as np

income = np.array([20000, 25000, 30000, 35000, 1000000])

median_income = np.median(income)

print("Median Income:", median_income)
```

## Interpretation

### Output

Median Income: 30000.0

### What does the program do?

The program calculates the middle value of the income data using np.median().

### What statistical concept does it demonstrate?

It demonstrates the median as a measure of central tendency.

### How do we interpret the output?

The median income is ₹30,000.

The extreme value of ₹10,00,000 does not change the median significantly.

### Where is it used in AI/ML?

Median is useful for understanding skewed data and can also be used to fill missing numerical values when the data contains outliers.

# Topic 3 – Mode

## Step 1 – Understand the Concept

Mode is the value that occurs most frequently in a dataset.

A dataset can have:
- One mode – Unimodal
- Two modes – Bimodal
- More than two modes – Multimodal
- No repeated value – No mode

There is no mathematical formula for mode.

Mode = Most frequently occurring value

## Step 2 – Demonstrate the Concept

### Numerical Example

Data:

10, 20, 20, 30, 40, 20, 50

The value 20 appears three times.

Therefore:

Mode = 20

### Business Example

A clothing store records the shirt sizes sold:

M, L, M, S, M, L, XL

The size M occurs most frequently.

Therefore, M is the mode.

The store can use this information to understand which shirt size is most commonly purchased.

### AI/ML Example

Suppose a dataset contains customer payment methods:

UPI, Card, UPI, Cash, UPI, Card

UPI occurs most frequently.

Therefore, UPI is the mode.

Mode is useful for finding the most common category in categorical data.

### How does this demonstrate the concept?

The mode identifies the value or category that appears most frequently in the dataset


## Step 3: Implement the Concept

```python
from statistics import mode

sizes = ["M", "L", "M", "S", "M", "L", "XL"]

most_common_size = mode(sizes)

print("Most Common Size:", most_common_size)
```

## Interpretation

### Output

Most Common Size: M

### What does the program do?

The program finds the value that occurs most frequently in the list.

### What statistical concept does it demonstrate?

It demonstrates mode.

### How do we interpret the output?

M is the most frequently sold shirt size.

### Where is it used in AI/ML?

Mode can be used to understand categorical data and can also be used for filling missing categorical values.

# Topic 4 – Weighted Mean

## Step 1 – Understand the Concept

Weighted mean is an average where different values have different levels of importance.

In a normal mean, every value has equal importance.

In a weighted mean, each value is multiplied by its corresponding weight.

### Formula

Weighted Mean = Σ(X × W) / ΣW

Where:

X = Value

W = Weight

The sum of the weighted values is divided by the total weight.

## Step 2 – Demonstrate the Concept

### Numerical Example

Suppose a student's final grade is calculated using:

Assignment = 80, Weight = 20%
Midterm = 70, Weight = 30%
Final Exam = 90, Weight = 50%

Weighted Mean:

= (80 × 0.20) + (70 × 0.30) + (90 × 0.50)

= 16 + 21 + 45

= 82

Therefore, the final weighted score is 82.

### Business Example

A company evaluates an employee using:

Project Performance = 80, Weight = 30%
Quality = 90, Weight = 20%
Productivity = 85, Weight = 50%

Productivity has a higher weight, so it has a greater effect on the final performance score.

### AI/ML Example

Weighted averages can be used when some observations should have greater importance than others.

For example, a business may give more importance to recent customer interactions when calculating a customer score.

### How does this demonstrate the concept?

The values do not contribute equally. Each value contributes according to its assigned weight.


## Step 3: Implement the Concept

```python
import numpy as np

scores = np.array([80, 70, 90])
weights = np.array([0.20, 0.30, 0.50])

weighted_mean = np.average(scores, weights=weights)

print("Weighted Mean:", weighted_mean)
```

## Interpretation

### Output

Weighted Mean: 82.0

### What does the program do?

The program calculates the weighted average of the scores using their corresponding weights.

### What statistical concept does it demonstrate?

It demonstrates weighted mean.

### How do we interpret the output?

The final weighted score is 82.

The final exam has the highest weight, so it has the greatest effect on the final score.

### Where is it used in AI/ML?

Weighted averages can be used when different observations or data points need different levels of importance.

# Topic 5 – Geometric Mean

## Step 1 – Understand the Concept

Geometric mean is an average used mainly for values that are multiplied together, such as growth rates, ratios, and investment returns.

Instead of adding the values and dividing them, we multiply all the values and then take the appropriate root.

### Formula

Geometric Mean = ⁿ√(X₁ × X₂ × ... × Xₙ)

Where:

n = Number of values

X = Each value

Geometric mean is especially useful when dealing with compound or multiplicative changes.

## Step 2 – Demonstrate the Concept

### Numerical Example

Suppose an investment has growth factors:

1.10, 1.20, 1.30

Geometric Mean:

= ³√(1.10 × 1.20 × 1.30)

≈ 1.197

Therefore, the average growth factor is approximately 1.197.

### Business Example

Suppose a company has different yearly growth rates.

Because growth compounds from one year to the next, geometric mean is more appropriate for finding the average growth factor than simply taking an arithmetic average.

### AI/ML Example

Geometric mean can be used when evaluating metrics or ratios where multiplicative changes are important.

It is useful in situations where values combine proportionally rather than additively.

### How does this demonstrate the concept?

The values are multiplied together before taking the root, which makes geometric mean suitable for growth and multiplicative data.


## Step 3: Implement the Concept

```python
import numpy as np

growth_factors = np.array([1.10, 1.20, 1.30])

geometric_mean = np.prod(growth_factors) ** (1 / len(growth_factors))

print("Geometric Mean:", geometric_mean)
```

## Interpretation

### Output

Geometric Mean: 1.197...

### What does the program do?

The program multiplies all growth factors and calculates the appropriate root.

### What statistical concept does it demonstrate?

It demonstrates geometric mean.

### How do we interpret the output?

The average multiplicative growth factor is approximately 1.197.

### Where is it used in AI/ML?

Geometric mean can be used for growth-related metrics, ratios, and situations where values combine multiplicatively.

# Topic 6 – Harmonic Mean

## Step 1 – Understand the Concept

Harmonic mean is a type of average that is especially useful for rates and ratios.

It is commonly used when the same quantity is involved at different rates.

### Formula

Harmonic Mean = n / Σ(1/X)

For two values:

H = 2ab / (a + b)

Where:

n = Number of values

X = Each value

## Step 2 – Demonstrate the Concept

### Numerical Example

Suppose a car travels the same distance at:

60 km/h going to a destination
40 km/h returning from the destination

Harmonic Mean:

H = (2 × 60 × 40) / (60 + 40)

H = 4800 / 100

H = 48 km/h

Therefore, the average speed is 48 km/h.

### Business Example

Harmonic mean can be used when averaging rates, such as productivity rates or certain performance ratios.

### AI/ML Example

The F1-score in classification uses the harmonic mean of precision and recall.

F1 = 2 × (Precision × Recall) / (Precision + Recall)

It gives a balanced measure when both precision and recall are important.

### How does this demonstrate the concept?

The harmonic mean is appropriate for averaging rates and ratios because it gives more importance to smaller values.


## Step 3: Implement the Concept

```python
%pip install scipy
from scipy.stats import hmean

speeds = [60, 40]

harmonic_mean = hmean(speeds)

print("Harmonic Mean:", harmonic_mean)
```

## Interpretation

### Output

Harmonic Mean: 48.0

### What does the program do?

The program calculates the harmonic mean of two speeds using scipy.stats.hmean().

### What statistical concept does it demonstrate?

It demonstrates harmonic mean for averaging rates.

### How do we interpret the output?

The harmonic mean of 60 km/h and 40 km/h is 48 km/h.

For equal distances, this is the appropriate average speed.

### Where is it used in AI/ML?

Harmonic mean is directly used in the F1-score, which combines precision and recall into one evaluation metric.
