# 1. Sampling

## Step 1 – Understand the Concept

### Definition

Sampling is the process of selecting a smaller group of observations from a larger population for analysis.

Instead of collecting data from the entire population, we select a sample that can provide useful information about the population.

### Why is Sampling Important?

- Saves time
- Reduces cost
- Makes data collection easier
- Useful when studying the entire population is difficult

### Key Terms

- Population → Entire group being studied
- Sample → Smaller group selected from the population

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose a college has 5,000 students.

Instead of asking all 5,000 students about their satisfaction, we select 500 students and collect their responses.

Population = 5,000 students
Sample = 500 students

The 500 selected students are used to understand the overall student population.

### Numerical Example

Population = 10,000
Sample = 500

Sampling Percentage:

(500 / 10,000) × 100 = 5%

Therefore, 5% of the population is selected as the sample.

### Business Example

An e-commerce company has 100,000 customers.

The company selects 2,000 customers to understand customer satisfaction instead of contacting all customers.

### AI/ML Example

When a dataset contains millions of records, a smaller representative sample can be used for initial data exploration and testing.


## Step 3 – Implement the Concept

```python
import numpy as np

population = np.arange(1, 10001)

sample = np.random.choice(
    population,
    size=500,
    replace=False
)

print("Population Size:", len(population))
print("Sample Size:", len(sample))
print("First 10 Sample Values:", sample[:10])
```

## Interpretation of Results

The program creates a population of 10,000 observations and randomly selects 500 observations as the sample.

The sample represents 5% of the population.

### Advantages

- Saves time and resources.
- Easier to analyze.
- Useful for large populations.

### Limitations

- A sample may not perfectly represent the population.
- Poor sampling methods can introduce bias.

### Interview Questions and Answers

#### 1. What is Sampling?
Sampling is the process of selecting a smaller group from a larger population for analysis.

#### 2. What is a Population?
A population is the complete group being studied.

#### 3. What is a Sample?
A sample is a smaller group selected from the population.

#### 4. Why do we use Sampling?
We use sampling to save time, cost, and effort.

#### 5. Give an AI/ML example.
A representative subset of a large dataset can be selected for initial analysis or experimentation.

### Self Reflection

I learned that sampling helps us study a large population by selecting a smaller and manageable group of observations.

# 2. Random Sampling

## Step 1 – Understand the Concept

### Definition

Random Sampling is a method where members of a population are selected randomly.

Each member has an equal or known chance of being selected.

The selection is not based on personal preference.

### Where is it Used?

- Surveys
- Research
- Customer feedback
- Data analysis
- AI/ML datasets

## Step 2 – Demonstrate the Concept

### Real-World Example

A college has 1,000 students.

A computer randomly selects 100 students for a survey.

Every student has a chance of being selected.

### Numerical Example

Population = 1,000
Sample = 100

Sampling Percentage:

(100 / 1,000) × 100 = 10%

### Business Example

A company randomly selects 500 customers from its customer database to collect feedback.

### AI/ML Example

A random subset of a large dataset can be selected for model experimentation or testing.


## Step 3 – Implement the Concept

```python
import numpy as np

population = np.arange(1, 1001)

random_sample = np.random.choice(
    population,
    size=100,
    replace=False
)

print("Population Size:", len(population))
print("Sample Size:", len(random_sample))
print("Random Sample:", random_sample[:10])
```

## Interpretation of Results

The program randomly selects 100 observations from a population of 1,000.

The observations are selected without intentionally choosing specific members.

### Advantages

- Simple to understand.
- Reduces selection bias when properly applied.
- Easy to implement.

### Limitations

- Requires access to the population list or sampling frame.
- A random sample can still differ from the population by chance.

### Interview Questions and Answers

#### 1. What is Random Sampling?
It is a method of selecting observations randomly from a population.

#### 2. Does every member have a chance to be selected?
Yes, under simple random sampling every member has an equal chance.

#### 3. What is one advantage?
It reduces intentional selection bias.

#### 4. Where is it used?
It is used in surveys, research, and dataset sampling.

#### 5. How can Python perform random sampling?
NumPy's `random.choice()` can be used.

### Self Reflection

I learned how random sampling selects observations without intentionally preferring specific members of the population.

# 3. Stratified Sampling

## Step 1 – Understand the Concept

### Definition

Stratified Sampling divides a population into smaller groups called strata based on a specific characteristic.

Then, samples are selected from each stratum.

Examples of strata include:

- Department
- Age group
- Location
- Education level
- Customer type

### Why is it Used?

It helps ensure that important groups within the population are represented in the sample.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company has employees from three departments:

IT = 500
HR = 200
Finance = 300

Instead of selecting employees from the entire company without considering departments, we select employees from each department.

### Numerical Example

Suppose the population contains:

IT = 50%
HR = 20%
Finance = 30%

For a sample of 100 employees:

IT sample = 50
HR sample = 20
Finance sample = 30

### Business Example

An e-commerce company has:

- Premium customers
- Regular customers
- New customers

The company can select customers from each group.

### AI/ML Example

If a dataset contains different customer segments, stratified sampling can help maintain representation from each segment.


## Step 3 – Implement the Concept

```python
import pandas as pd

data = pd.DataFrame({
    "Employee": range(1, 11),
    "Department": [
        "IT", "IT", "IT", "IT",
        "HR", "HR",
        "Finance", "Finance", "Finance", "Finance"
    ]
})

sample = data.groupby(
    "Department",
    group_keys=False
).sample(
    frac=0.5,
    random_state=42
)

print(sample)
```

## Interpretation of Results

The population is divided into departments.

The program selects observations separately from each department.

Therefore, IT, HR, and Finance are represented in the sample.

### Advantages

- Ensures important groups are represented.
- Useful when population groups are different.
- Can improve sample representation.

### Limitations

- Requires information about population groups.
- More complex than simple random sampling.

### Interview Questions and Answers

#### 1. What is Stratified Sampling?
It divides the population into groups and selects samples from each group.

#### 2. What are the groups called?
They are called strata.

#### 3. Why is Stratified Sampling used?
To ensure important groups are represented.

#### 4. Give a business example.
Selecting customers separately from premium, regular, and new customer groups.

#### 5. What is one limitation?
Information about the population groups is required.

### Self Reflection

I learned how stratified sampling divides a population into meaningful groups and selects observations from each group.

# 4. Cluster Sampling

## Step 1 – Understand the Concept

### Definition

Cluster Sampling divides a population into groups called clusters.

Instead of selecting individual observations from every cluster, some clusters are randomly selected for the study.

### Example

A country can be divided into:

- Cities
- Districts
- Villages

Researchers can randomly select some cities and collect data from those selected cities.

### Numerical Example

### Limitations

- Selected clusters may not represent the entire population.
- Sampling error can be higher if clusters are very different.

# 5. Systematic Sampling

## Step 1 – Understand the Concept

### Definition

Systematic Sampling selects observations from a population at a regular interval.

First, a starting point is selected. Then every kth observation is selected.

### Formula

k = N / n

Where:

- N = population size
- n = required sample size
- k = sampling interval

## Step 2 – Demonstrate the Concept

### Numerical Example

Population size = 1,000
Sample size = 100

k = 1,000 / 100
k = 10

Therefore, every 10th observation can be selected after choosing a starting point.

### Real-World Example

A supermarket wants to survey customers.

It can randomly choose the first customer and then select every 10th customer.

### Business Example

A manufacturing company can inspect every 20th product coming from a production line.

### AI/ML Example

Systematic sampling can be used to create a regularly spaced sample from ordered data.


## Step 3 – Implement the Concept

```python
import numpy as np

population = np.arange(1, 101)

sample_size = 10

interval = len(population) // sample_size

start = np.random.randint(0, interval)

systematic_sample = population[start::interval][:sample_size]

print("Sampling Interval:", interval)
print("Systematic Sample:", systematic_sample)
```

## Interpretation of Results

The program calculates a sampling interval and selects observations at that regular interval.

### Advantages

- Simple and easy to implement.
- Useful for ordered populations.
- Less time-consuming.

### Limitations

- Periodic patterns in the data can create bias.
- Requires an ordered population.

### Interview Questions and Answers

#### 1. What is Systematic Sampling?
It selects observations at a fixed interval.

#### 2. What is the sampling interval?
k = N / n.

#### 3. What does N represent?
Population size.

#### 4. What does n represent?
Required sample size.

#### 5. Give a business example.
Inspecting every 20th product on a production line.

### Self Reflection

I learned how systematic sampling uses a fixed interval to select observations from an ordered population.

# 6. Sampling Bias

## Step 1 – Understand the Concept

### Definition

Sampling Bias occurs when the selected sample does not properly represent the population.

This can happen when some members have a higher or lower chance of being selected.

### Common Causes

- Convenience sampling
- Self-selection
- Undercoverage
- Non-response
- Poor sampling method

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose a college wants to know whether all students are satisfied.

If the survey is given only to students who attend one particular event, the sample may not represent all students.

### Numerical Example

Population:

Group A = 500
Group B = 500

But the sample contains:

Group A = 90
Group B = 10

The population has equal representation, but the sample is heavily dominated by Group A.

This can introduce sampling bias.

### Business Example

An online company asks only its most active customers for feedback.

The results may not represent inactive customers.

### AI/ML Example

If training data overrepresents one customer group and underrepresents another, an ML model may learn patterns that do not represent the broader population.


## Step 3 – Implement the Concept

```python
import numpy as np

population = np.array(["A"] * 500 + ["B"] * 500)

biased_sample = np.array(["A"] * 90 + ["B"] * 10)

print("Population:")
print("Group A:", np.sum(population == "A"))
print("Group B:", np.sum(population == "B"))

print("\nSample:")
print("Group A:", np.sum(biased_sample == "A"))
print("Group B:", np.sum(biased_sample == "B"))
```

## Interpretation of Results

The population has equal representation from Group A and Group B.

However, the sample contains 90% Group A and only 10% Group B.

Therefore, the sample does not represent the population well and may produce biased conclusions.

### Advantages

Understanding sampling bias helps us design better sampling methods and datasets.

### Limitations

Bias can be difficult to identify when complete population information is unavailable.

### Interview Questions and Answers

#### 1. What is Sampling Bias?
It occurs when a sample does not properly represent the population.

#### 2. What causes sampling bias?
Poor selection methods, non-response, undercoverage, and self-selection can cause it.

#### 3. Why is it important in AI/ML?
Biased training data can lead to unreliable model behavior.

#### 4. Give a business example.
Asking only highly active customers for feedback.

#### 5. How can sampling bias be reduced?
Use appropriate sampling methods and try to obtain a representative sample.

### Self Reflection

I learned that a poorly selected sample can misrepresent the population and lead to incorrect conclusions.

# 7. Sampling Error

## Step 1 – Understand the Concept

### Definition

Sampling Error is the difference between a population value and the corresponding value calculated from a sample.

It occurs because we study only a sample instead of the complete population.

### Formula

Sampling Error = Sample Statistic - Population Parameter

For example:

Sampling Error = Sample Mean - Population Mean

## Step 2 – Demonstrate the Concept

### Numerical Example

Population Mean = 50

Sample Mean = 48

Sampling Error:

48 - 50 = -2

Absolute Sampling Error:

|-2| = 2

### Real-World Example

Suppose the average height of all students in a college is 165 cm.

A selected sample may have an average height of 163 cm.

The difference between 163 cm and 165 cm is sampling error.

### Business Example

Actual average customer spending = ₹1,000

Sample average spending = ₹950

Sampling Error:

₹950 - ₹1,000 = -₹50

### AI/ML Example

A sample used to estimate the average value of a feature may produce a value different from the actual population average.


## Step 3 – Implement the Concept

```python
import numpy as np

population = np.arange(1, 101)

population_mean = np.mean(population)

sample = np.random.choice(
    population,
    size=20,
    replace=False
)

sample_mean = np.mean(sample)

sampling_error = sample_mean - population_mean

print("Population Mean:", population_mean)
print("Sample Mean:", sample_mean)
print("Sampling Error:", sampling_error)
```

## Interpretation of Results

The sample mean may differ from the population mean because only a subset of observations was selected.

A smaller sampling error means the sample estimate is closer to the population value.

### Advantages

- Helps us understand the difference between sample estimates and population values.
- Useful for evaluating statistical estimates.

### Limitations

- Sampling error cannot be completely eliminated when only a sample is used.
- The error can vary from one sample to another.

### Interview Questions and Answers

#### 1. What is Sampling Error?
It is the difference between a sample statistic and the corresponding population parameter.

#### 2. Why does it occur?
Because we use a sample instead of the complete population.

#### 3. Can sampling error be zero?
Yes, it can be zero for a particular sample, but it is not guaranteed.

#### 4. How can sampling error generally be reduced?
Using a larger and appropriately selected sample can reduce sampling variability.

#### 5. Give a business example.
The difference between actual average customer spending and sample average spending.

### Self Reflection

I learned that sample estimates can differ from population values and that this difference is called sampling error.

# 8. Sample Size

## Step 1 – Understand the Concept

### Definition

Sample Size is the number of observations selected from a population for a study.

Choosing an appropriate sample size is important because:

- A very small sample may give unreliable results.
- A larger sample generally provides more information.
- A very large sample may require more time and resources.

### Formula

For estimating a population proportion:

n = Z² × p × (1-p) / E²

Where:

- n = required sample size
- Z = z-score for the confidence level
- p = estimated population proportion
- E = margin of error

## Step 2 – Demonstrate the Concept

### Numerical Example

Suppose:

Confidence Level = 95%
Z = 1.96
p = 0.5
Margin of Error = 0.05

Formula:

n = (1.96² × 0.5 × 0.5) / (0.05²)

n ≈ 384.16

Since we cannot collect a fraction of an observation:

Required Sample Size ≈ 385

### Real-World Example

A company wants to survey customers with a 95% confidence level and a 5% margin of error.

Instead of asking every customer, it can determine an appropriate sample size.

### Business Example

An e-commerce company with thousands of customers can calculate how many customers should be surveyed to estimate customer satisfaction.

### AI/ML Example

Sample size is important when collecting training, validation, and evaluation data. A sufficiently large and representative dataset can provide more reliable model evaluation.


## Step 3 – Implement the Concept

```python
import math

Z = 1.96
p = 0.5
E = 0.05

sample_size = (Z ** 2 * p * (1 - p)) / (E ** 2)

print("Calculated Sample Size:", sample_size)
print("Required Sample Size:", math.ceil(sample_size))
```

## Interpretation of Results

The calculated sample size is approximately 384.16.

Since a fraction of an observation cannot be collected, we round up.

Required Sample Size = 385

Therefore, approximately 385 observations are required under these assumptions.

### Advantages

- Helps plan data collection.
- Helps balance accuracy and resources.
- Avoids unnecessarily small samples.

### Limitations

- The required size depends on assumptions such as confidence level and margin of error.
- A large sample does not automatically remove sampling bias.

### Interview Questions and Answers

#### 1. What is Sample Size?
It is the number of observations included in a sample.

#### 2. Why is sample size important?
It affects the reliability and precision of statistical estimates.

#### 3. What happens with a very small sample?
Results may have higher sampling variability.

#### 4. What does margin of error represent?
It represents the desired maximum amount of estimation error.

#### 5. What is the common formula for a population proportion?
n = Z²p(1-p)/E²

I learned how sample size is determined and how confidence level, population proportion, and margin of error affect the required number of observations.
