# Topic 1 – T-Test

## Step 1 – Understand the Concept

A T-Test is a statistical test used to determine whether there is a statistically significant difference between means.

It is commonly used when the population standard deviation is unknown and the sample size is relatively small.

There are different types of T-Tests:

1. One-Sample T-Test
2. Independent Two-Sample T-Test
3. Paired T-Test

In this example, I will use a One-Sample T-Test.

The basic idea is to compare a sample mean with a hypothesized population mean.

### Formula

t = (x̄ - μ₀) / (s / √n)

Where:

- x̄ = Sample mean
- μ₀ = Hypothesized population mean
- s = Sample standard deviation
- n = Sample size

## Assumptions

A T-Test generally assumes:

1. The observations are independent.
2. The variable is numerical.
3. The data is approximately normally distributed, especially for small samples.
4. For an independent two-sample T-Test, the standard version assumes equal variances; Welch's T-Test can be used when equal variance is not reasonable.

## When to Use

Use a T-Test when:

- Comparing a sample mean with a known/hypothesized value.
- Comparing means between two groups.
- Comparing before-and-after measurements for the same subjects.
- Population standard deviation is unknown.

## When Not to Use

Do not use a T-Test when:

- The observations are not independent when independence is required.
- The data is strongly non-normal for a small sample.
- The variable is categorical.
- There are more than two independent groups; ANOVA may be more appropriate.

For heavily non-normal data, a non-parametric test may be considered depending on the research question.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company wants to know whether the average delivery time is different from the target of 30 minutes.

H₀: μ = 30

Hₐ: μ ≠ 30

A one-sample T-Test can be used.

### Numerical Example

Suppose the delivery times are:

28, 31, 29, 32, 30, 27, 31, 29

We can compare the sample mean with 30 minutes.

### Business Example

A company wants to determine whether the average production time after introducing a new process differs from the previous target.

### AI/ML Example

A T-Test can be used in an experiment to compare a model's performance against a baseline when the observations and experimental design satisfy the test assumptions.


## Step 3 – Implement the Concept

```python
import numpy as np
from scipy.stats import ttest_1samp

delivery_time = np.array([28, 31, 29, 32, 30, 27, 31, 29])

target = 30

t_statistic, p_value = ttest_1samp(delivery_time, target)

print("T-statistic:", t_statistic)
print("P-value:", p_value)
```

## Interpretation

The T-Test compares the sample mean with the hypothesized mean of 30 minutes.

If the p-value is less than the selected significance level, such as 0.05, we reject the null hypothesis.

If the p-value is greater than or equal to 0.05, we fail to reject the null hypothesis.

The result should be interpreted along with the assumptions and practical importance.

## Advantages

- Simple and widely used.
- Useful for comparing means.
- Available in SciPy.

## Limitations

- Sensitive to violations of assumptions, especially with small samples.
- Not suitable for categorical variables.
- Extreme outliers can affect the result.

# Topic 2 – Z-Test

## Step 1 – Understand the Concept

A Z-Test is a statistical test used to compare a sample statistic with a population parameter when the population standard deviation is known or when the large-sample conditions justify a normal approximation.

For a one-sample mean Z-Test:

z = (x̄ - μ₀) / (σ / √n)

Where:

- x̄ = Sample mean
- μ₀ = Hypothesized population mean
- σ = Population standard deviation
- n = Sample size

The Z-Test uses the standard normal distribution.

## Assumptions

A Z-Test generally assumes:

1. Observations are independent.
2. The variable is numerical.
3. Population standard deviation is known for the classical one-sample mean Z-Test.
4. The population is normally distributed, or the sample is sufficiently large for a normal approximation to be reasonable.

## When to Use

Use a Z-Test when:

- Population standard deviation is known.
- The sample is sufficiently large for the required approximation.
- The observations are independent.
- The variable is numerical.

## When Not to Use

Do not use a classical Z-Test when:

- Population standard deviation is unknown and a T-Test is more appropriate.
- The observations are dependent.
- The required distributional or large-sample assumptions are not reasonable.

## Step 2 – Demonstrate the Concept

### Real-World Example

A factory knows that the population standard deviation of bottle filling volume is 5 ml.

The company wants to test whether the average filling volume is 500 ml.

### Numerical Example

Suppose:

Sample mean = 502 ml

Population mean = 500 ml

Population standard deviation = 5 ml

Sample size = 100

The Z-Test can determine whether the observed difference is statistically significant.

### Business Example

A manufacturer can test whether the average product weight differs from the specified target when the population standard deviation is known.

### AI/ML Example

Z-based statistical comparisons can be used in large-sample experiments when the normal approximation is justified.


## Step 3 – Implement the Concept

```python
import numpy as np
from scipy.stats import norm

sample_mean = 502
population_mean = 500
population_std = 5
sample_size = 100

z_statistic = (
    sample_mean - population_mean
) / (population_std / np.sqrt(sample_size))

p_value = 2 * (1 - norm.cdf(abs(z_statistic)))

print("Z-statistic:", z_statistic)
print("P-value:", p_value)
```

## Interpretation

The Z-statistic tells us how many population standard errors the sample mean is away from the hypothesized mean.

The p-value is compared with the chosen significance level.

If p-value < 0.05, we reject H₀.

If p-value >= 0.05, we fail to reject H₀.

## Advantages

- Simple for large-sample or known-standard-deviation situations.
- Uses the standard normal distribution.
- Easy to calculate.

## Limitations

- Classical one-sample Z-Test requires the population standard deviation to be known.
- Sensitive to incorrect assumptions.
- Often a T-Test is more appropriate when σ is unknown.

# Topic 3 – Chi-Square Test

## Step 1 – Understand the Concept

The Chi-Square Test is used for categorical data.

Two common forms are:

1. Chi-Square Test of Independence
2. Chi-Square Goodness-of-Fit Test

The Chi-Square Test of Independence checks whether two categorical variables are associated.

For example:

- Gender and product preference
- Payment method and customer type
- Device type and subscription status

### Formula

χ² = Σ [(Observed - Expected)² / Expected]

Where:

- Observed = Actual frequency
- Expected = Expected frequency

## Assumptions

For a Chi-Square Test of Independence:

1. Variables should be categorical.
2. Observations should be independent.
3. Data should be represented as frequency counts.
4. Expected cell frequencies should generally be sufficiently large for the usual chi-square approximation.

## When to Use

Use it when:

- Both variables are categorical.
- You want to test whether two categorical variables are associated.
- Data is represented using counts/frequencies.

## When Not to Use

Do not use it when:

- The variables are continuous without appropriate categorization.
- Observations are dependent.
- Expected frequencies are too small for the usual approximation.

For small expected counts, an exact test such as Fisher's exact test may be more appropriate for suitable 2×2 tables.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company wants to know whether customer subscription type is related to device preference.

Variables:

- Subscription: Free / Premium
- Device: Mobile / Desktop

Both variables are categorical.

### Numerical Example

Suppose the company collects customer counts:

|              | Mobile | Desktop |
|--------------|--------|---------|
| Free         | 40     | 20      |
| Premium      | 30     | 50      |

A Chi-Square Test can determine whether subscription type and device preference are associated.

### Business Example

An e-commerce company can test whether payment method is associated with customer age group.

### AI/ML Example

Chi-Square tests can be used in feature selection for categorical features to assess their association with a categorical target.


## Step 3 – Implement the Concept

```python
import numpy as np
from scipy.stats import chi2_contingency

data = np.array([
    [40, 20],
    [30, 50]
])

chi2, p_value, degrees_of_freedom, expected = chi2_contingency(data)

print("Chi-Square Statistic:", chi2)
print("P-value:", p_value)
print("Degrees of Freedom:", degrees_of_freedom)

print("\nExpected Frequencies:")
print(expected)
```

## Interpretation

The test checks whether the two categorical variables are independent.

If p-value < 0.05, we reject the null hypothesis of independence and conclude that there is statistical evidence of an association between the variables.

If p-value >= 0.05, we fail to reject the null hypothesis.

A statistically significant association does not by itself establish causation.

## Advantages

- Useful for categorical data.
- Easy to implement.
- Useful for testing associations.

## Limitations

- Does not directly measure the strength of association.
- Requires appropriate expected frequencies.
- Statistical significance does not imply causation.

# Topic 4 – ANOVA

## Step 1 – Understand the Concept

ANOVA stands for Analysis of Variance.

It is used to test whether the means of three or more groups are statistically different.

Instead of performing many separate T-Tests, ANOVA provides an overall test.

### Hypotheses

H₀:

All group means are equal.

Hₐ:

At least one group mean is different.

### Basic Idea

ANOVA compares:

Between-group variation

with

Within-group variation.

The test statistic is:

F = Between-Group Variance / Within-Group Variance

## Assumptions

One-way ANOVA generally assumes:

1. Observations are independent.
2. The dependent variable is numerical.
3. The groups are independent.
4. The residuals are approximately normally distributed within groups.
5. Group variances are reasonably similar for the classical ANOVA procedure.

## When to Use

Use ANOVA when:

- Comparing means across three or more independent groups.
- The dependent variable is numerical.
- The assumptions are reasonably satisfied.

## When Not to Use

Do not use classical one-way ANOVA when:

- The observations are dependent.
- The outcome is categorical.
- The assumptions are severely violated.

For non-normal data or unequal variances, alternatives such as Kruskal-Wallis or Welch's ANOVA may be considered depending on the design.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company wants to compare average sales generated by three different marketing strategies.

Groups:

- Strategy A
- Strategy B
- Strategy C

ANOVA can test whether at least one average sales value differs.

### Numerical Example

Sales:

Strategy A: 20, 22, 21, 19, 23

Strategy B: 30, 32, 29, 31, 33

Strategy C: 25, 26, 24, 27, 25

ANOVA can compare all three group means together.

### Business Example

An organization can compare average employee productivity across three different training programs.

### AI/ML Example

ANOVA can be used in experimental analysis to compare model performance across multiple groups or experimental conditions when its assumptions and experimental design are appropriate.


## Step 3 – Implement the Concept

```python
from scipy.stats import f_oneway

strategy_a = [20, 22, 21, 19, 23]
strategy_b = [30, 32, 29, 31, 33]
strategy_c = [25, 26, 24, 27, 25]

f_statistic, p_value = f_oneway(
    strategy_a,
    strategy_b,
    strategy_c
)

print("F-statistic:", f_statistic)
print("P-value:", p_value)
```

## Interpretation

The ANOVA test checks the null hypothesis that all group means are equal.

If p-value < 0.05, we reject H₀ and conclude that at least one group mean differs.

ANOVA itself does not tell us exactly which groups differ.

If the overall ANOVA is significant, post-hoc tests such as Tukey's HSD can be used to investigate pairwise differences while controlling the appropriate error rate.

## Advantages

- Compares three or more means in one overall test.
- Avoids performing many individual T-Tests.
- Useful in experiments and business analysis.

## Limitations

- Does not identify which specific groups differ.
- Depends on assumptions such as independence and variance structure.
- Multiple comparisons require appropriate post-hoc procedures.

# Topic 5 – Mann-Whitney U Test

## Step 1 – Understand the Concept

The Mann-Whitney U Test is a non-parametric test used to compare two independent groups.

It is an alternative to the independent two-sample T-Test when the assumptions for the T-Test are not appropriate.

The test works with the ranks of observations rather than directly comparing the raw values.

It is useful when the data is ordinal or when the distributional assumptions of a parametric test are not reasonable.

The test evaluates whether the distributions of the two groups differ.

Under common conditions, it is often described as testing whether one group tends to have larger values than the other.

## Assumptions

The Mann-Whitney U Test generally assumes:

1. The two groups are independent.
2. Observations within each group are independent.
3. The outcome is at least ordinal.
4. The observations can be ranked.

## When to Use

Use it when:

- Comparing two independent groups.
- Data is ordinal.
- Data is not reasonably modeled by normal distributions.
- A non-parametric alternative is appropriate.

## When Not to Use

Do not use it when:

- The groups are paired or dependent.
- There are more than two independent groups.
- The observations are not independent.

For paired data, the Wilcoxon signed-rank test may be more appropriate.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company wants to compare customer satisfaction scores between two branches.

Branch A:

7, 8, 6, 9, 7

Branch B:

5, 6, 4, 5, 6

The scores can be treated as ordinal ratings.

### Numerical Example

The test ranks all observations and compares the rank distributions between the two groups.

### Business Example

An e-commerce company can compare customer satisfaction ratings for two different delivery services.

### AI/ML Example

The Mann-Whitney U Test can be used to compare model performance across two independent experimental groups when the performance measurements do not satisfy the assumptions required for a parametric comparison.


## Step 3 – Implement the Concept

```python
from scipy.stats import mannwhitneyu

group_a = [7, 8, 6, 9, 7]
group_b = [5, 6, 4, 5, 6]

u_statistic, p_value = mannwhitneyu(
    group_a,
    group_b,
    alternative="two-sided"
)

print("U-statistic:", u_statistic)
print("P-value:", p_value)
```

## Interpretation

The test compares the rank distributions of the two independent groups.

If p-value < 0.05, we reject the null hypothesis and conclude that there is statistical evidence of a difference between the groups under the test framework.

The Mann-Whitney U Test should not automatically be interpreted as a test of means.

## Advantages

- Non-parametric.
- Does not require normality of the groups.
- Useful for ordinal data.
- Less sensitive to extreme values than mean-based tests in some situations.

## Limitations

- It does not directly test a difference in means.
- Interpretation can be more complicated when the group distributions have different shapes.
- Requires independent observations.

# Topic 6 – Shapiro-Wilk Test

## Step 1 – Understand the Concept

The Shapiro-Wilk Test is a statistical test used to assess whether a sample is consistent with a normal distribution.

### Hypotheses

H₀:

The data is consistent with a normal distribution.

Hₐ:

The data is not consistent with a normal distribution.

A small p-value provides evidence against normality.

The Shapiro-Wilk test should be used together with visual methods such as histograms and Q-Q plots, especially because statistical tests can be sensitive to sample size.

## Assumptions

The Shapiro-Wilk test is designed for numerical continuous data and tests the normality assumption.

Observations should generally be independent.

## When to Use

Use it when:

- You need to assess normality of a sample.
- You are checking assumptions for a statistical procedure.
- The dataset is reasonably sized for the test.

## When Not to Use

Do not rely on it alone when:

- The sample is extremely large, because very small deviations from normality may become statistically significant.
- The sample is extremely small, because the test may have limited ability to detect non-normality.
- The data is categorical.

Visual checks such as Q-Q plots should complement the test.

## Step 2 – Demonstrate the Concept

### Real-World Example

A researcher wants to use a T-Test and first wants to investigate whether the sample data is reasonably consistent with normality.

### Numerical Example

Suppose we have employee productivity values:

50, 52, 49, 51, 53, 48, 50, 52, 51, 49

The Shapiro-Wilk Test can be used to assess whether the sample is consistent with a normal distribution.

### Business Example

Before applying a parametric test to customer spending data, a company can investigate the distributional assumptions.

### AI/ML Example

During EDA and statistical analysis, normality tests can help understand the distribution of numerical features or residuals.

However, Machine Learning algorithms do not generally require every feature to be normally distributed.


## Step 3 – Implement the Concept

```python
from scipy.stats import shapiro

data = [50, 52, 49, 51, 53, 48, 50, 52, 51, 49]

statistic, p_value = shapiro(data)

print("Shapiro-Wilk Statistic:", statistic)
print("P-value:", p_value)
```

## Interpretation

The null hypothesis states that the data is consistent with a normal distribution.

If p-value < 0.05, we reject H₀ and conclude that the sample provides evidence against normality.

If p-value >= 0.05, we fail to reject H₀.

Failing to reject H₀ does not prove that the data is perfectly normal.

A histogram and Q-Q plot should also be considered.

## Advantages

- Simple normality test.
- Available in SciPy.
- Useful for checking assumptions.

## Limitations

- Sensitive to sample size.
- A significant result does not describe how severe the deviation is.
- Should not be used as the only evidence for normality.

# Interview Questions & Answers

## 1. What is a T-Test?

A T-Test is used to determine whether there is a statistically significant difference between means.

It is commonly used when the population standard deviation is unknown.

---

## 2. What is the difference between a T-Test and a Z-Test?

A T-Test is commonly used when the population standard deviation is unknown and the appropriate assumptions are satisfied.

A classical Z-Test for a population mean assumes the population standard deviation is known, or uses a large-sample normal approximation when justified.

---

## 3. What is a Chi-Square Test?

A Chi-Square Test is used with categorical data.

It can test whether two categorical variables are associated or whether observed frequencies fit an expected distribution.

---

## 4. What is ANOVA?

ANOVA stands for Analysis of Variance.

It is used to test whether the means of three or more independent groups are statistically different.

---

## 5. Why not perform multiple T-Tests instead of ANOVA?

Performing many separate T-Tests increases the overall probability of obtaining a false positive.

ANOVA provides an overall test for differences among multiple group means.

If ANOVA is significant, suitable post-hoc tests can be used to investigate specific group differences.

---

## 6. What is the Mann-Whitney U Test?

It is a non-parametric test used to compare two independent groups using ranked observations.

It can be used when the assumptions for a parametric two-sample comparison are not appropriate.

---

## 7. What is the Shapiro-Wilk Test?

The Shapiro-Wilk Test assesses whether a sample is consistent with a normal distribution.

A small p-value provides evidence against the normality assumption.

---

## 8. When would you use a Chi-Square Test instead of a T-Test?

A Chi-Square Test is used for categorical variables and frequency counts.

A T-Test is used for comparing means of numerical data.

---

## 9. When would you use ANOVA instead of a T-Test?

ANOVA is generally used when comparing three or more independent groups.

A T-Test is generally used for one mean or a comparison involving two groups, depending on the specific T-Test.

---

## 10. Does the Shapiro-Wilk Test prove that data is normal?

No.

Failing to reject the null hypothesis means there is not sufficient evidence against normality based on the test.

It does not prove that the data is perfectly normally distributed.

Visual methods such as Q-Q plots should also be considered.
