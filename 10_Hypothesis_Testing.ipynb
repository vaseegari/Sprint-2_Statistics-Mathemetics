# Topic 1 – Null Hypothesis

## Step 1 – Understand the Concept

The Null Hypothesis, written as H₀, is the initial assumption that there is no significant effect, difference, or relationship in the population.

In hypothesis testing, we start with the null hypothesis and use sample data to determine whether there is enough statistical evidence to reject it.

The null hypothesis does not necessarily mean that nothing exists. It represents the baseline or default assumption.

### Example

Suppose a company claims that the average delivery time is 30 minutes.

The null hypothesis can be:

H₀: μ = 30 minutes

This means the population mean delivery time is 30 minutes.

## Step 2 – Demonstrate the Concept

### Real-World Example

A school wants to check whether the average exam score of students is 70.

The null hypothesis can be:

H₀: μ = 70

### Numerical Example

Suppose a sample of students has an average score of 72.

We test whether this difference from 70 is statistically significant.

### Business Example

A company claims that its average delivery time is 30 minutes.

The null hypothesis assumes that the average delivery time is 30 minutes.

### AI/ML Example

In Machine Learning experiments, a null hypothesis can represent the assumption that a new model or feature does not produce a statistically significant improvement compared with a baseline.

## Step 3 – Implement the Concept

```python
import numpy as np
from scipy import stats

sample = np.array([68, 72, 70, 75, 69, 71, 73, 70, 74, 68])

sample_mean = np.mean(sample)

print("Sample Mean:", sample_mean)
```

## Interpretation

The sample mean is calculated to compare the sample with the value stated in the null hypothesis.

The sample mean alone is not enough to reject the null hypothesis.

We need a statistical test and a p-value to determine whether there is sufficient evidence against H₀.

## Advantages

- Provides a clear baseline for statistical testing.
- Helps structure a statistical investigation.
- Can be tested using different statistical methods.

## Limitations

- Failing to reject H₀ does not prove that H₀ is true.
- The conclusion depends on the sample, test, assumptions, and significance level.

# Topic 2 – Alternative Hypothesis

## Step 1 – Understand the Concept

The Alternative Hypothesis, written as H₁ or Hₐ, represents the claim that there is a significant difference, effect, or relationship.

It is the hypothesis that we consider when the sample data provides sufficient evidence against the null hypothesis.

For example:

H₀: μ = 30

Hₐ: μ ≠ 30

This means we are testing whether the population mean is different from 30.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company wants to know whether a new delivery system changes the average delivery time.

H₀: Average delivery time = 30 minutes

Hₐ: Average delivery time ≠ 30 minutes

### Numerical Example

If the sample mean is 35 minutes, we can test whether the difference from 30 minutes is statistically significant.

### Business Example

A company introduces a new advertisement and wants to know whether average sales have changed.

H₀: New advertisement does not change average sales.

Hₐ: New advertisement changes average sales.

### AI/ML Example

Suppose a new Machine Learning model is compared with an existing model.

H₀: There is no significant difference in performance.

Hₐ: There is a significant difference in performance.

## Interpretation

The alternative hypothesis represents the effect or difference we are trying to find evidence for.

The exact form depends on the research question.

Examples:

Hₐ: μ > 30 → Greater than 30

Hₐ: μ < 30 → Less than 30

Hₐ: μ ≠ 30 → Different from 30

## Advantages

- Clearly defines the effect being investigated.
- Helps determine whether a test should be one-tailed or two-tailed.

## Limitations

- The alternative hypothesis does not automatically become true just because H₀ is rejected.
- The hypothesis must be defined before analyzing the results to avoid biased conclusions.

# Topic 3 – p-value

## Step 1 – Understand the Concept

A p-value measures how compatible the observed sample result is with the null hypothesis, under the assumptions of the statistical test.

A small p-value means the observed result would be relatively unusual if the null hypothesis were true.

A common decision rule is:

If p-value < α:
    Reject H₀

If p-value ≥ α:
    Fail to reject H₀

The p-value does NOT tell us the probability that the null hypothesis is true.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company wants to determine whether a new training program changes employee productivity.

A very small p-value provides statistical evidence against the assumption that there is no change.

### Numerical Example

Suppose:

p-value = 0.03

Significance level α = 0.05

Since:

0.03 < 0.05

we reject H₀.

### Business Example

A company tests whether a new marketing campaign changes average sales.

If p-value = 0.02 and α = 0.05, the result provides statistical evidence against the null hypothesis.

### AI/ML Example

Suppose we compare two models using a suitable statistical test.

A p-value can help assess whether the observed performance difference is statistically significant under the test assumptions.


## Step 3 – Implement the Concept

```python
import numpy as np
from scipy.stats import ttest_1samp

sample = np.array([68, 72, 70, 75, 69, 71, 73, 70, 74, 68])

population_mean = 70

t_statistic, p_value = ttest_1samp(sample, population_mean)

print("T-statistic:", t_statistic)
print("P-value:", p_value)
```

## Interpretation

The t-test compares the sample mean with the hypothesized population mean of 70.

The p-value is compared with the selected significance level.

For example, if:

p-value < 0.05

we reject H₀.

If:

p-value ≥ 0.05

we fail to reject H₀.

The exact conclusion should be based on the calculated p-value and the assumptions of the test.

## Advantages

- Provides a quantitative way to assess evidence against H₀.
- Commonly used in statistical testing.
- Available through statistical libraries such as SciPy.

## Limitations

- A p-value does not measure effect size.
- A small p-value does not prove practical importance.
- Interpretation depends on the test assumptions and study design.

# Topic 4 – Confidence Interval

## Step 1 – Understand the Concept

A confidence interval gives a range of plausible values for a population parameter based on sample data.

For example, a 95% confidence interval gives a method that, over repeated sampling under the same conditions, would capture the true population parameter about 95% of the time.

For a population mean, a common form is:

Confidence Interval = Sample Mean ± Critical Value × Standard Error

A confidence interval provides more information than a single point estimate because it shows the uncertainty around the estimate.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company estimates that the average customer waiting time is 15 minutes.

Instead of reporting only 15 minutes, it may report a confidence interval such as:

12 to 18 minutes.

This provides information about the uncertainty of the estimate.

### Numerical Example

Suppose:

Sample Mean = 50

95% Confidence Interval = [47, 53]

The interval gives a range of plausible values for the population mean under the method used.

### Business Example

A business can estimate average customer spending using a confidence interval.

### AI/ML Example

Confidence intervals can be used to express uncertainty around estimated model performance or statistical parameters when the appropriate assumptions and evaluation procedure are satisfied.


## Step 3 – Implement the Concept

```python
import numpy as np
from scipy import stats

sample = np.array([68, 72, 70, 75, 69, 71, 73, 70, 74, 68])

confidence_level = 0.95

mean = np.mean(sample)
standard_error = stats.sem(sample)

confidence_interval = stats.t.interval(
    confidence_level,
    df=len(sample) - 1,
    loc=mean,
    scale=standard_error
)

print("Sample Mean:", mean)
print("95% Confidence Interval:", confidence_interval)
```

## Interpretation

The confidence interval provides a range of plausible values for the population mean based on the sample and the assumptions of the method.

A narrower interval generally indicates greater precision, while a wider interval indicates greater uncertainty.

## Advantages

- Shows uncertainty around an estimate.
- More informative than a single point estimate.
- Useful for statistical reporting and decision-making.

## Limitations

- Interpretation depends on the sampling method and statistical assumptions.
- A 95% confidence interval does not mean there is a 95% probability that this particular fixed parameter lies inside the calculated interval.

# Topic 5 – Significance Level

## Step 1 – Understand the Concept

The significance level is represented by α (alpha).

It is the threshold used to decide when the evidence against the null hypothesis is strong enough to reject H₀.

Common significance levels are:

- 0.10
- 0.05
- 0.01

The most commonly used level is 0.05.

For example:

α = 0.05

If p-value < 0.05, we reject H₀.

If p-value ≥ 0.05, we fail to reject H₀.

The significance level should ideally be selected before examining the test result.

## Step 2 – Demonstrate the Concept

### Real-World Example

A researcher chooses α = 0.05 before conducting a study.

This means the test uses a 5% significance threshold.

### Numerical Example

Suppose:

p-value = 0.03

α = 0.05

Since 0.03 < 0.05, H₀ is rejected.

### Business Example

A company tests whether a new process changes average production time and chooses α = 0.05 before analyzing the result.

### AI/ML Example

When statistically comparing model results, a predefined significance level can be used with an appropriate statistical test to evaluate whether an observed difference is statistically significant.

## Interpretation

The significance level determines the threshold for statistical decision-making.

It should not be changed simply because the obtained p-value is inconvenient.

## Advantages

- Provides a predefined decision threshold.
- Helps control the probability of Type I error.

## Limitations

- Statistical significance does not necessarily mean practical importance.
- The choice of α affects the testing procedure.
- A 0.05 threshold is conventional, not a universal requirement.

# Topic 6 – Type I Error

## Step 1 – Understand the Concept

A Type I Error occurs when we reject the null hypothesis even though the null hypothesis is actually true.

It is also called a **false positive**.

Example:

H₀: A new medicine has no effect.

If we reject H₀ and conclude that the medicine has an effect when it actually does not, this is a Type I Error.

The significance level α represents the maximum Type I error rate targeted by the test procedure under its assumptions.

## Step 2 – Demonstrate the Concept

### Real-World Example

A security system incorrectly identifies a normal person as a threat.

This is similar to a false positive.

### Numerical Example

If α = 0.05, the test procedure is designed to have a 5% Type I error rate under the null hypothesis, assuming the test conditions are satisfied.

### Business Example

A company concludes that a new marketing campaign increased sales when there was actually no real increase.

### AI/ML Example

In a classification problem, a false positive occurs when a negative case is incorrectly classified as positive.

For example, a normal transaction is incorrectly classified as fraudulent.

## Interpretation

Type I Error means:

**Rejecting a true null hypothesis.**

It is commonly associated with a false positive.

## Advantages

Understanding Type I Error helps researchers control false-positive conclusions.

## Limitations

Reducing Type I Error by choosing a smaller significance level can make it harder to detect real effects and may increase the chance of Type II Error.

# Topic 7 – Type II Error

## Step 1 – Understand the Concept

A Type II Error occurs when we fail to reject the null hypothesis even though the null hypothesis is actually false.

It is also called a **false negative**.

Example:

H₀: A new training program has no effect.

If the training program actually improves performance but the test fails to detect the improvement, this is a Type II Error.

The probability of a Type II Error is commonly represented by β (beta).

Statistical power is:

Power = 1 - β

## Step 2 – Demonstrate the Concept

### Real-World Example

A medical test fails to detect a disease that a patient actually has.

This is similar to a false negative.

### Numerical Example

Suppose a test has:

β = 0.20

Then:

Power = 1 - 0.20 = 0.80

So the statistical power is 80%.

### Business Example

A company introduces a genuinely effective marketing campaign, but the statistical test fails to detect the improvement.

### AI/ML Example

A fraud detection system may fail to identify an actually fraudulent transaction.

This is analogous to a false negative in classification.

## Interpretation

Type II Error means:

**Failing to reject a false null hypothesis.**

It is commonly associated with a false negative.

## Advantages

Understanding Type II Error helps researchers design studies with adequate statistical power.

## Limitations

- Reducing Type II Error generally requires enough sample information and an appropriate study design.
- A failure to reject H₀ does not prove that there is no effect.

# Topic 8 – One-Tailed Test

## Step 1 – Understand the Concept

A one-tailed test is used when the alternative hypothesis specifies a direction.

The alternative hypothesis may be:

Hₐ: μ > μ₀

or:

Hₐ: μ < μ₀

Examples:

- Testing whether average sales increased.
- Testing whether a new method reduced processing time.

A one-tailed test places the rejection region in one direction of the sampling distribution.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company introduces a new system and wants to test whether it reduces average processing time.

H₀: μ ≥ 30

Hₐ: μ < 30

This is a left-tailed test.

### Numerical Example

Suppose the target mean is 30 and the researcher only wants to know whether the new process reduces the mean below 30.

### Business Example

A company wants to determine whether a new advertising strategy increases sales.

Hₐ: μ > μ₀

This is a right-tailed test.

### AI/ML Example

If a new optimization method is specifically expected to improve model accuracy compared with a baseline, a directional statistical test may be considered if its assumptions and experimental design support it.



## Step 3 – Implement the Concept

```python
import numpy as np
from scipy.stats import ttest_1samp

sample = np.array([25, 27, 26, 24, 25, 23, 26, 24, 25, 24])

target_mean = 30

t_statistic, two_tailed_p = ttest_1samp(sample, target_mean)

# Convert the two-sided p-value to a one-sided p-value
# because the alternative is sample mean < target_mean.
if t_statistic < 0:
    one_tailed_p = two_tailed_p / 2
else:
    one_tailed_p = 1 - (two_tailed_p / 2)

print("T-statistic:", t_statistic)
print("One-tailed p-value:", one_tailed_p)
```

## Interpretation

The alternative hypothesis in this example is that the population mean is less than 30.

Therefore, we use a one-tailed test.

The p-value is interpreted using the selected significance level.

The direction of the hypothesis must be decided before looking at the result.

## Advantages

- More statistical power in the specified direction when a directional hypothesis is justified.
- Useful when only one direction is scientifically or practically relevant.

## Limitations

- Cannot be used to make a claim about the opposite direction.
- Choosing a one-tailed test after seeing the data can lead to misleading conclusions.

# Topic 9 – Two-Tailed Test

## Step 1 – Understand the Concept

A two-tailed test is used when we want to determine whether a parameter is different from a specified value in either direction.

The alternative hypothesis is:

Hₐ: μ ≠ μ₀

It can detect both:

- An increase
- A decrease

The rejection regions are located in both tails of the sampling distribution.

## Step 2 – Demonstrate the Concept

### Real-World Example

A company wants to know whether a new process changes the average delivery time.

It does not assume beforehand whether delivery time will increase or decrease.

H₀: μ = 30

Hₐ: μ ≠ 30

Therefore, a two-tailed test is appropriate.

### Numerical Example

Suppose:

Target mean = 30

Sample mean = 35

The test checks whether the difference is statistically significant, regardless of whether the true mean is above or below 30.

### Business Example

A company wants to determine whether a price change affects sales, without specifying the direction beforehand.

### AI/ML Example

When comparing two models, researchers may test whether their performance differs in either direction rather than assuming beforehand which model will perform better.


## Step 3 – Implement the Concept

```python
import numpy as np
from scipy.stats import ttest_1samp

sample = np.array([32, 31, 35, 33, 30, 34, 32, 31, 36, 33])

target_mean = 30

t_statistic, p_value = ttest_1samp(sample, target_mean)

print("T-statistic:", t_statistic)
print("Two-tailed p-value:", p_value)
```

## Interpretation

The alternative hypothesis is:

Hₐ: μ ≠ 30

Therefore, this is a two-tailed test.

The p-value is compared with the chosen significance level.

If p-value < α, we reject H₀.

If p-value ≥ α, we fail to reject H₀.

## Advantages

- Detects statistically significant differences in either direction.
- Appropriate when the research question does not specify a direction.

## Limitations

- Requires evidence to be sufficiently extreme in either direction.
- Less concentrated in one direction than a one-tailed test.

# 5+ Interview Questions & Answers

## 1. What is hypothesis testing?

Hypothesis testing is a statistical method used to evaluate a claim about a population using sample data.

It starts with a null hypothesis and uses statistical evidence to determine whether there is sufficient evidence against it.

---

## 2. What is the difference between Null and Alternative Hypothesis?

The Null Hypothesis (H₀) represents the baseline assumption, such as no difference or no effect.

The Alternative Hypothesis (Hₐ) represents the difference, effect, or relationship being investigated.

---

## 3. What is a p-value?

A p-value measures how compatible the observed data is with the null hypothesis, under the assumptions of the statistical test.

A small p-value indicates stronger evidence against H₀.

---

## 4. What is significance level?

The significance level, α, is the threshold used for deciding when evidence against H₀ is statistically significant.

A commonly used value is 0.05.

If p-value < α, we reject H₀.

---

## 5. What is Type I Error?

Type I Error occurs when we reject a true null hypothesis.

It is commonly called a false positive.

---

## 6. What is Type II Error?

Type II Error occurs when we fail to reject a false null hypothesis.

It is commonly called a false negative.

---

## 7. What is the difference between one-tailed and two-tailed tests?

A one-tailed test checks a directional alternative such as:

Hₐ: μ > μ₀

or:

Hₐ: μ < μ₀

A two-tailed test checks whether the parameter is different in either direction:

Hₐ: μ ≠ μ₀

---

## 8. What is a confidence interval?

A confidence interval provides a range of plausible values for a population parameter based on sample data and the assumptions of the statistical method.

It provides information about the uncertainty of an estimate.

---

## 9. Does a p-value tell us the probability that H₀ is true?

No.

A p-value is not the probability that H₀ is true.

It measures how compatible the observed result is with H₀ under the assumptions of the test.

---

## 10. What does "fail to reject H₀" mean?

It means the sample does not provide sufficient statistical evidence to reject the null hypothesis at the selected significance level.

It does not mean that H₀ has been proven true.
