# Topic 1 – Data Cleaning

## Step 1 – Understand the Concept

Data cleaning is the process of identifying and correcting problems
in a dataset before using it for analysis or Machine Learning.

Statistical techniques help us understand whether data contains
unusual values, incorrect values, duplicates, inconsistent formats,
or unexpected distributions.

Common statistical techniques used during data cleaning include:

- Mean
- Median
- Mode
- Standard deviation
- Quartiles
- Minimum and maximum
- Distribution analysis

### Purpose

The purpose of data cleaning is to improve data quality so that the
Machine Learning model receives reliable input.

### When to Use

Data cleaning is required before most Machine Learning projects,
especially when the raw data comes from real-world systems.

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose a dataset contains customer ages:

22, 25, 27, 30, 500

The value 500 is probably incorrect for a normal customer-age
dataset.

Statistical analysis can help identify this unusual value.

### Numerical Example

Consider:

10, 12, 11, 13, 12, 100

The value 100 is much larger than the other observations.

Mean and standard deviation can help us investigate the unusual
observation.

### Business Example

An e-commerce company may have incorrect product prices because
of data-entry errors.

Statistical summaries can help identify unusual prices.

### AI/ML Example

If incorrect values are given to a Machine Learning model,
the model may learn misleading patterns.

Therefore, statistical analysis is performed before model training.

- Mean

## Advantages

- Improves data quality.
- Helps identify unusual values.

## Limitations

  is actually wrong.

# Topic 2 – Missing Value Handling

## Step 1 – Understand the Concept

Missing values occur when some observations do not contain a value.

Statistics can help us decide how to handle missing values.

Common approaches include:

- Mean imputation
- Median imputation
- Mode imputation
- Removing observations
- Using model-based methods

Mean is commonly used when data is reasonably symmetric.

Median is often preferred when data contains skewness or outliers.

Mode can be used for categorical data.

### Purpose

The purpose is to handle missing information without introducing
unnecessary bias into the dataset.

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose some customers did not provide their age during registration.

Instead of immediately deleting those customers, we can investigate
the distribution of available ages.

### Numerical Example

Ages:

20, 22, 24, NaN, 26, 28

The median or mean of the available values can be used to fill the
missing value depending on the data distribution.

### Business Example

A bank may have missing customer-income values.

Deleting every customer with a missing income could reduce the
available training data.

### AI/ML Example

Many Machine Learning algorithms cannot directly work with missing
numerical values.

Therefore, missing values often need to be handled before model
training.


## Step 3 – Implement the Concept

```python
import pandas as pd

data = pd.DataFrame({
    "Age": [20, 22, 24, None, 26, 28]
})

median_age = data["Age"].median()

data["Age"] = data["Age"].fillna(median_age)

print(data)
```



### What Does the Program Do?

The program calculates the median age and replaces the missing
value with the median.

### Statistical Concept Demonstrated

The program demonstrates median-based imputation.

### How to Interpret the Output

The missing age is replaced by the median of the available ages.

### AI/ML Usage

Median imputation is commonly used for numerical features when the
data may contain skewness or outliers.

## Advantages

- Simple.
- Easy to implement.
- Keeps the number of observations unchanged.

## Limitations

- It may reduce natural variability.
- It assumes the chosen statistic is suitable for the data.
- It may not be appropriate when missingness has a meaningful pattern.

# Topic 3 – Outlier Detection

## Step 1 – Understand the Concept

An outlier is an observation that is unusually different from the
majority of observations.

Statistical methods commonly used for outlier detection include:

- IQR method
- Z-score method
- Box plot
- Standard deviation

### Purpose

The purpose is to identify unusual observations that may affect
analysis or Machine Learning models.

### When to Use

Outlier detection is useful when:

- Data contains extreme values.
- A model is sensitive to unusual observations.
- We want to investigate data-quality issues.

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose most employee salaries are between ₹20,000 and ₹80,000,
but one record shows ₹10,00,000.

This value should be investigated.

### Numerical Example

Values:

10, 12, 11, 13, 12, 100

100 is much larger than the other observations.

### Business Example

A credit-card transaction dataset may contain a very large
transaction compared with normal customer transactions.

### AI/ML Example

Extreme values can influence models such as Linear Regression,
so detecting them can be important during preprocessing.

## Step 3 – Implement the Concept

```python
import numpy as np

data = np.array([10, 12, 11, 13, 12, 100])

mean = np.mean(data)
std = np.std(data)

z_scores = (data - mean) / std

print("Z-Scores:", z_scores)
```



### What Does the Program Do?

The program calculates the mean and standard deviation and then
calculates a Z-score for every observation.

### Statistical Concept Demonstrated

A Z-score indicates how far an observation is from the mean in
terms of standard deviations.

### How to Interpret the Output

An observation with a very large positive or negative Z-score may
be considered unusual.

The result should be investigated rather than automatically deleted.

### AI/ML Usage

Outlier detection is useful during preprocessing because extreme
values can influence some Machine Learning algorithms.

## Advantages

- Simple mathematical approach.
- Useful for approximately normally distributed data.

## Limitations

- Can be affected by existing outliers.
- Z-score assumptions may not work well for highly skewed data.

# Topic 5 – Feature Scaling

## Step 1 – Understand the Concept

Feature scaling changes numerical features to comparable scales.

This is important when different features have very different
ranges.

For example:

Age: 20–60

Salary: 20,000–1,00,000

Without scaling, algorithms based on distance or magnitude may be
strongly influenced by larger numerical values.

Common methods include:

- Standardization
- Min-Max Scaling

### Standardization

A common formula is:

z = (x - mean) / standard deviation

### Purpose

The purpose is to make numerical features comparable in scale.

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose a model uses:

Age
Income

Income has much larger numerical values than age.

Scaling puts them on a comparable scale.

### Numerical Example

If:

Mean = 50
Standard deviation = 10
Value = 70

Then:

z = (70 - 50) / 10
z = 2

### Business Example

A customer segmentation model may use income and number of
website visits.

Scaling prevents income from dominating distance calculations
simply because it has larger numerical values.

### AI/ML Example

Scaling is important for algorithms such as:

- K-Means
- K-Nearest Neighbors
- Support Vector Machines
- PCA



## Step 3 – Implement the Concept

```python
import pandas as pd

data = pd.DataFrame({
    "Age": [20, 30, 40, 50, 60]
})

mean = data["Age"].mean()
std = data["Age"].std()

data["Standardized_Age"] = (
    (data["Age"] - mean) / std
)

print(data)
```



### What Does the Program Do?

The program calculates the mean and standard deviation of age
and converts each age into a standardized value.

### Statistical Concept Demonstrated

It demonstrates standardization using the Z-score formula.

### How to Interpret the Output

Values above zero are above the mean.

Values below zero are below the mean.

A value of approximately 2 means the observation is around
two standard deviations above the mean.

### AI/ML Usage

Scaling is especially important for distance-based and
variance-sensitive algorithms.

## Advantages

- Makes features comparable.
- Helps distance-based algorithms.
- Useful for PCA.

## Limitations

- Standardization does not remove outliers.
- Scaling should be learned from training data when building
  predictive models.

# Topic 6 – Data Normalization

## Step 1 – Understand the Concept

Data normalization transforms numerical values to a common range.

One common method is Min-Max normalization.

### Formula

x_normalized = (x - minimum) / (maximum - minimum)

This commonly transforms values into the range 0 to 1.

### Purpose

The purpose is to bring numerical values into a comparable range.

### When to Use

Normalization can be useful when:

- Features have different ranges.
- Distance calculations are involved.
- The algorithm benefits from bounded values.

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose a dataset contains:

Age: 20–60
Income: 20,000–1,00,000

Normalization converts both variables to a similar scale.

### Numerical Example

Minimum = 10
Maximum = 50
Value = 30

Normalized value:

(30 - 10) / (50 - 10)
= 20 / 40
= 0.5

### Business Example

Customer features such as spending and visits can have different
ranges.

Normalization makes their numerical scales comparable.

### AI/ML Example

Normalization can be useful for algorithms that use distances,
such as K-Means and KNN.


## Step 3 – Implement the Concept

```python
import numpy as np

data = np.array([10, 20, 30, 40, 50])

normalized = (
    (data - data.min()) /
    (data.max() - data.min())
)

print(normalized)
```



### What Does the Program Do?

The program applies Min-Max normalization to the data.

### Statistical Concept Demonstrated

It transforms the minimum value to 0 and the maximum value to 1.

### How to Interpret the Output

All values are now between 0 and 1.

The relative order of the values is maintained.

### AI/ML Usage

Normalization can help distance-based algorithms work with features
that originally have different scales.

## Advantages

- Simple.
- Produces a bounded range.
- Useful for distance-based algorithms.

## Limitations

- Sensitive to extreme minimum and maximum values.
- The chosen range may not be appropriate for every algorithm.

# Topic 8 – Linear Regression

## Step 1 – Understand the Concept

Linear Regression is a supervised Machine Learning algorithm used
to predict a continuous numerical target.

It estimates a relationship between input variables and a target
variable.

A simple linear regression equation is:

y = b0 + b1x

Where:

y  = predicted value
b0 = intercept
b1 = slope
x  = input feature

Statistics is important because Linear Regression estimates the
relationship between variables and analyzes prediction errors.

### Purpose

The purpose is to predict continuous numerical values.

### Example Applications

- House price prediction
- Sales prediction
- Salary prediction
- Demand forecasting

## Step 2 – Demonstrate the Concept

### Real-World Example

House price may be related to house area.

Generally, larger houses may have higher prices, although many
other factors can also influence price.

### Numerical Example

If:

House Area = 1000 sq.ft
Price = ₹50 lakh

and larger areas are associated with higher prices, regression can
estimate the relationship.

### Business Example

A company can use advertising expenditure to predict sales.

### AI/ML Example

Linear Regression learns coefficients that minimize prediction
error between actual and predicted values.


## Step 3 – Implement the Concept

```python
import numpy as np

x = np.array([1, 2, 3, 4, 5])
y = np.array([2, 4, 5, 8, 10])

slope, intercept = np.polyfit(x, y, 1)

predicted = slope * x + intercept

print("Slope:", slope)
print("Intercept:", intercept)
print("Predicted values:", predicted)
```



### What Does the Program Do?

The program calculates the best-fitting straight line for the
given observations using least squares.

### Statistical Concept Demonstrated

It demonstrates estimation of a linear relationship between an
input and a continuous target.

### How to Interpret the Output

The slope indicates how much the predicted target changes for a
one-unit change in the input.

The intercept represents the estimated target when x is zero.

### AI/ML Usage

Linear Regression is useful for continuous prediction problems.

# Topic 10 – Naive Bayes

## Step 1 – Understand the Concept

Naive Bayes is a probabilistic Machine Learning algorithm based
on Bayes' theorem.

It calculates the probability of a class based on observed
features.

The "naive" assumption is that features are conditionally
independent given the class.

Bayes' theorem is:

P(A|B) = P(B|A)P(A) / P(B)

### Purpose

The purpose is to classify observations using probability.

### Common Applications

- Spam detection
- Text classification
- Sentiment classification
- Document classification

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose an email contains words such as:

"Win"
"Prize"
"Free"

A spam classifier can use the probability of these words occurring
in spam emails to estimate whether the email is spam.

### Numerical Example

Suppose:

P(Spam) = 0.4

The probability of observing specific words given spam can then
be used to update the probability of the email being spam.

### Business Example

An email provider can classify incoming emails as spam or
not spam.

### AI/ML Example

Naive Bayes is particularly useful for text classification because
word frequencies can be converted into probabilistic features.


## Step 3 – Implement the Concept

```python
from math import prod

p_spam = 0.4
p_not_spam = 0.6

word_probs_spam = [0.8, 0.7]
word_probs_not_spam = [0.1, 0.2]

spam_score = p_spam * prod(word_probs_spam)
not_spam_score = p_not_spam * prod(word_probs_not_spam)

total = spam_score + not_spam_score

spam_probability = spam_score / total

print("Probability of Spam:", spam_probability)
```



### What Does the Program Do?

The program calculates a simplified Naive Bayes probability
using a prior probability and feature likelihoods.

### Statistical Concept Demonstrated

It demonstrates probabilistic classification using Bayes' theorem.

### How to Interpret the Output

The calculated value represents the estimated probability that
the observation belongs to the spam class based on the simplified
probabilities.

### AI/ML Usage

Naive Bayes is commonly used for text classification and spam
detection.

## Advantages

- Simple and fast.
- Works well for many text-classification problems.
- Requires relatively little training data.

## Limitations

- The feature independence assumption may not hold in real data.
- Probability estimates may not always be well calibrated.

# Topic 11 – Clustering

## Step 1 – Understand the Concept

Clustering is an unsupervised Machine Learning technique that
groups similar observations together.

Statistics helps us understand:

- Distance
- Mean
- Variance
- Distribution
- Cluster characteristics

One common clustering algorithm is K-Means.

K-Means assigns observations to clusters based on their distance
from cluster centroids.

### Purpose

The purpose is to discover natural groups in unlabeled data.

### Business Applications

- Customer segmentation
- Product grouping
- Market segmentation
- Document grouping

## Step 2 – Demonstrate the Concept

### Real-World Example

A company may want to group customers into:

- Low spenders
- Medium spenders
- High spenders

without manually assigning labels.

### Numerical Example

Customers can be represented using:

Annual Income
Annual Spending

Customers with similar values can be grouped together.

### Business Example

An e-commerce company can segment customers based on purchasing
behavior.

### AI/ML Example

K-Means can identify groups in unlabeled datasets.


## Step 3 – Implement the Concept

```python
import numpy as np

data = np.array([
    [1, 1],
    [1, 2],
    [2, 1],
    [8, 8],
    [9, 8],
    [8, 9]
])

centroid_1 = np.array([1, 1])
centroid_2 = np.array([8, 8])

for point in data:
    distance_1 = np.linalg.norm(point - centroid_1)
    distance_2 = np.linalg.norm(point - centroid_2)

    if distance_1 < distance_2:
        cluster = 1
    else:
        cluster = 2

    print(point, "-> Cluster", cluster)
```



### What Does the Program Do?

The program calculates the distance between each observation and
two predefined cluster centers.

Each observation is assigned to the nearest center.

### Statistical Concept Demonstrated

It demonstrates distance-based grouping.

### How to Interpret the Output

Points close to `[1, 1]` are assigned to Cluster 1.

Points close to `[8, 8]` are assigned to Cluster 2.

### AI/ML Usage

Clustering is used for customer segmentation, recommendation,
pattern discovery, and exploratory analysis.

## Advantages

- Does not require labeled data.
- Useful for discovering groups.
- Simple concept.

## Limitations

- The number of clusters may need to be chosen.
- Sensitive to scaling and outliers.
- Different initial centroids can affect results.

# Topic 12 – Principal Component Analysis (PCA)

## Step 1 – Understand the Concept

PCA is a dimensionality-reduction technique.

It transforms a dataset containing correlated variables into a
smaller number of new variables called principal components.

The first principal component captures the largest amount of
variance possible.

The second component captures the next largest amount of variance
while being uncorrelated with the first component.

### Purpose

PCA is used to:

- Reduce dimensionality.
- Remove redundant information.
- Visualize high-dimensional data.
- Improve computational efficiency.

### Statistical Concepts Used

PCA depends heavily on:

- Mean
- Variance
- Covariance
- Correlation
- Eigenvectors
- Eigenvalues

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose a dataset contains 20 features describing customers.

Many features may contain overlapping information.

PCA can transform these features into fewer principal components.

### Numerical Example

Suppose five original features explain information that can be
represented reasonably well using two principal components.

Instead of processing all five features, a model can work with
the two components.

### Business Example

A company may have many correlated customer behavior variables.

PCA can reduce them into a smaller set of components for analysis.

### AI/ML Example

PCA can reduce the number of features before training a Machine
Learning model.

It can also help visualize high-dimensional data in two dimensions.


## Step 3 – Implement the Concept

```python
import numpy as np

data = np.array([
    [2, 4],
    [3, 6],
    [4, 8],
    [5, 10],
    [6, 12]
])

centered_data = data - np.mean(data, axis=0)

covariance_matrix = np.cov(
    centered_data,
    rowvar=False
)

eigenvalues, eigenvectors = np.linalg.eig(
    covariance_matrix
)

print("Covariance Matrix:")
print(covariance_matrix)

print("\nEigenvalues:")
print(eigenvalues)

print("\nEigenvectors:")
print(eigenvectors)
```



### What Does the Program Do?

The program first centers the data by subtracting the mean.

Then it calculates the covariance matrix.

Finally, it calculates eigenvalues and eigenvectors.

### Statistical Concept Demonstrated

PCA uses covariance and variance to identify directions that
capture the largest amount of variation in the dataset.

### How to Interpret the Output

Larger eigenvalues represent directions containing more variance.

The corresponding eigenvectors represent the principal directions.

### AI/ML Usage

PCA can reduce dimensionality and help visualize complex datasets.

## Advantages

- Reduces the number of features.
- Can remove redundant information.
- Helps visualize high-dimensional data.
- Can reduce computational complexity.

## Limitations

- Principal components may be difficult to interpret.
- PCA is sensitive to feature scaling.
- Some information is lost when dimensions are removed.

## Step 2 – Demonstrate the Concept

### Real-World Example

Suppose two users have rated several movies similarly.

If User A likes a movie that User B has not watched, the system

### Numerical Example

User A ratings:

Movie 1 = 5
Movie 2 = 4
Movie 3 = 5

User B ratings:

Movie 1 = 5

### Business Example

## Step 3 – Implement the Concept

```python
import pandas as pd

ratings = pd.DataFrame({
    "Movie_1": [5, 5, 1],
    "Movie_2": [4, 4, 2],
    "Movie_3": [5, 4, 5],
    "Movie_4": [1, 2, 5]
}, index=["User_A", "User_B", "User_C"])

correlation = ratings.T.corr()

print("User Similarity:")
print(correlation)
```

rating patterns.

### Statistical Concept Demonstrated

- Can personalize recommendations.
- Uses historical user behavior.

## 1. Why is statistics important in Machine Learning?

Statistics helps us understand data, identify patterns, handle
missing values and outliers, select useful features, build models,
and evaluate model performance.

---

## 2. What is the difference between normalization and standardization?

Normalization commonly transforms values into a fixed range such
as 0 to 1.

Standardization transforms values based on the mean and standard
deviation so that the transformed data generally has mean 0 and
standard deviation close to 1.

---

## 3. Why is median often used for missing values?

Median is less affected by extreme values than the mean.

Therefore, it can be useful when numerical data is skewed or
contains outliers.

---

## 4. Why is feature scaling important?

Feature scaling makes numerical features comparable in magnitude.

It is especially important for algorithms that depend on distance
or variance, such as K-Means, KNN, SVM, and PCA.

---

## 5. What is an outlier?

An outlier is an observation that is unusually different from
the majority of observations.

It may represent an error, a rare event, or a genuine observation.

---

## 6. What is Linear Regression used for?

Linear Regression is used to predict a continuous numerical target
by estimating a relationship between input variables and the target.

---

## 7. What is Logistic Regression used for?

Logistic Regression is commonly used for classification.

It estimates probabilities that can be converted into class labels.

---

## 8. What is the main idea behind Naive Bayes?

Naive Bayes uses Bayes' theorem to calculate class probabilities
and assumes that features are conditionally independent given
the class.

---

## 9. Why is PCA used?

PCA is used to reduce dimensionality while retaining as much
variance as possible in fewer principal components.

---

## 10. What is clustering?

Clustering is an unsupervised Machine Learning technique that
groups similar observations together without predefined labels.

---

## 11. How is statistics used in recommendation systems?

Statistics can be used to analyze ratings, user behavior,
similarities, preferences, and interactions to generate
recommendations.

---

## 12. Why should we not automatically remove every outlier?

An outlier may be a genuine observation rather than an error.

For example, a very large transaction may represent a legitimate
purchase.

Therefore, the observation should be investigated before deciding
how to handle it.
